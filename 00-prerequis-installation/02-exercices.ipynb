## Comment travailler

Dans ce module, **tu ne tapes pas de code dans le notebook**. Pour chaque exercice :

1. Lis l'**objectif** et les **étapes**.
2. Tape les commandes **dans Git Bash** (ou le Terminal sur macOS/Linux). La cellule de code de l'exercice ne
   contient qu'un commentaire : tu n'as rien à y écrire.
3. Compare ce que le terminal affiche avec le **résultat attendu**.
4. Exécute la cellule **Vérification** (`Maj + Entrée`) : `✅ OK` = réussi ; sinon lis le message, il te dit quoi
   corriger.
5. Bloqué plus de 5 minutes ? Ouvre l'indice 💡, puis relis la section du cours indiquée.

**Si la vérification échoue** : un long message apparaît, surtout en anglais, qui se termine par une ligne comme
`AssertionError: ville vaut 'albi' au lieu de 'Albi'…`. `AssertionError` veut seulement dire **« la vérification a
échoué »**. Lis **uniquement le texte en français après les deux-points** : il te dit quoi corriger. Tout le reste
(lignes en anglais, numéros, flèches, lignes `assert …` de la cellule) peut être ignoré. Corrige ta cellule,
ré-exécute-la, puis ré-exécute la vérification.

**Caractères spéciaux** : les signes difficiles à taper (`°`, `É`, les pictogrammes comme ⛈ 🌧) se
**copient-collent depuis l'énoncé** : sélectionne-les à la souris, `Ctrl + C`, puis `Ctrl + V` dans ta cellule
(`Cmd` au lieu de `Ctrl` sur Mac). Au clavier français : `°` = `Maj + )` (Windows et Mac) ; `É` = `Verr. Maj` puis
`é` (Windows et Mac ; sur Windows, aussi `Alt + 0201` sur le pavé numérique). Recopie les textes **exactement**,
accents et espaces compris : la vérification compare au caractère près. Dans les formules, tape toi-même les
signes du clavier (`-`, `*`, `/`).

**Un terminal neuf à chaque exercice** : fais comme si chaque exercice commençait dans un terminal qu'on vient
d'ouvrir, dans ton dossier personnel, **sans environnement actif**. C'est pour ça que chaque exercice commence par
un `cd` (le plus souvent `cd ~/meteolocale/bac-a-sable`) : tape-le toujours, même si tu crois être déjà au bon
endroit. Il ne coûte rien et évite la plupart des erreurs.

**Où travailler** : tous les exercices se passent dans **`~/meteolocale/bac-a-sable/`**. Tu peux y faire toutes
les erreurs que tu veux. Si un dossier est raté, supprime-le avec l'explorateur de fichiers (ou le Finder) et
recommence.

**Activer un environnement** : la commande dépend de ton système. Dans les énoncés, « active
l'environnement » veut dire :
- 🪟 Windows (Git Bash) : `source .venv/Scripts/activate`
- 🍎 macOS / 🐧 Linux : `source .venv/bin/activate`

Les exercices s'enchaînent : fais-les **dans l'ordre**.

### 1.12 Créer un environnement virtuel

**Objectif** : Créer le dossier `projet-a` et, dedans, un environnement virtuel avec Python 3.13.

**À faire** :
1. Terminal neuf : commence par `cd ~/meteolocale/bac-a-sable` (même si tu crois y être déjà).
2. `mkdir projet-a`
3. `cd projet-a`
4. `uv venv --python 3.13`
5. `ls -a` : tu vois `.venv`.

**Résultat attendu** :
```
Using CPython 3.13.x
Creating virtual environment at: .venv
Activate with: source .venv/bin/activate
```

<details><summary>💡 Indice</summary>

`.venv` commence par un point : `ls` ne le montre pas, `ls -a` oui. Sur Windows, uv affiche `Activate with: .venv\Scripts\activate` : dans Git Bash, on tape `source .venv/Scripts/activate`. 📖 Cours, section « 10.3 L'environnement virtuel » ; guide, étape 5.

</details>

````python
# Rien à écrire ici : fais-le dans Git Bash, puis exécute la vérification.
````

````python
# Vérification
from pathlib import Path
import os
BASE = Path(os.environ.get("METEOLOCALE_BAC", Path.home() / "meteolocale" / "bac-a-sable"))

projet = BASE / "projet-a"
assert projet.is_dir(), f"Le dossier {projet} n'existe pas. Crée-le d'abord avec mkdir (depuis bac-a-sable)."
cfg = projet / ".venv" / "pyvenv.cfg"
if not cfg.exists() and (BASE / ".venv").exists():
    print("⚠️ Il y a un .venv directement dans bac-a-sable : tu as sans doute lancé uv venv avant d'entrer dans le dossier.")
assert cfg.exists(), f"Pas d'environnement virtuel dans {projet} (le fichier .venv/pyvenv.cfg est absent). Entre dans projet-a (cd projet-a) puis tape uv venv --python 3.13."

assert "3.13" in cfg.read_text(), "L'environnement n'utilise pas Python 3.13 : supprime le dossier .venv de projet-a et relance uv venv --python 3.13."
print("✅ OK")
````

### 1.13 Activer et garder la preuve

**Objectif** : Activer l'environnement de `projet-a`, puis écrire dans un fichier le chemin du Python utilisé.

**À faire** :
1. Terminal neuf : commence par `cd ~/meteolocale/bac-a-sable/projet-a` (même si tu crois y être déjà).
2. Active l'environnement : `source .venv/Scripts/activate` (🪟 Windows) ou `source .venv/bin/activate` (🍎 🐧).
3. Le prompt commence maintenant par `(projet-a)`.
4. `which python` : le chemin contient `projet-a/.venv`.
5. `which python > ou-est-python.txt` (la même réponse, écrite dans un fichier).

**Résultat attendu** :
```
(projet-a) …/bac-a-sable/projet-a $ which python
…/bac-a-sable/projet-a/.venv/bin/python
```

<details><summary>💡 Indice</summary>

Pas de `(projet-a)` au début du prompt ? La commande `source` a échoué : vérifie que tu es dans `projet-a` (`pwd`) et que tu utilises `Scripts` sur Windows, `bin` sur macOS/Linux. 📖 Cours, section « 10.3 L'environnement virtuel » ; guide, étape 5.

</details>

````python
# Rien à écrire ici : fais-le dans Git Bash, puis exécute la vérification.
````

````python
# Vérification
from pathlib import Path
import os
BASE = Path(os.environ.get("METEOLOCALE_BAC", Path.home() / "meteolocale" / "bac-a-sable"))

fichier = BASE / "projet-a/ou-est-python.txt"
assert fichier.is_file(), f"Je ne trouve pas le fichier {fichier}. Depuis projet-a, environnement activé, tape which python > ou-est-python.txt."
contenu = fichier.read_text(encoding="utf-8", errors="replace").strip().replace("\\", "/")

assert "projet-a/.venv" in contenu, f"Le fichier indique « {contenu} » : ce n'est pas le Python de projet-a/.venv. L'environnement n'était pas activé : vérifie que (projet-a) est au début du prompt, puis relance which python > ou-est-python.txt."
print("✅ OK")
````
