Comme dans les exercices : tape les commandes dans Git Bash (en commençant par `cd ~/meteolocale`), puis exécute la cellule
**Vérification** (`✅ OK` = réussi).

**Si la vérification échoue** : un long message apparaît, surtout en anglais, qui se termine par une ligne comme
`AssertionError: ville vaut 'albi' au lieu de 'Albi'…`. `AssertionError` veut seulement dire **« la vérification a
échoué »**. Lis **uniquement le texte en français après les deux-points** : il te dit quoi corriger. Tout le reste
(lignes en anglais, numéros, flèches, lignes `assert …` de la cellule) peut être ignoré. Corrige ta cellule,
ré-exécute-la, puis ré-exécute la vérification.

**Caractères spéciaux** : les signes difficiles à taper (`°`, `É`, les pictogrammes comme ⛈ 🌧) se
**copient-collent depuis l'énoncé** : sélectionne-les à la souris, `Ctrl + C`, puis `Ctrl + V` dans ta cellule
(`Cmd` au lieu de `Ctrl` sur Mac). Au clavier français : `°` = `Maj + )` (Windows et Mac) ; `É` = `Verr. Maj` puis
`é` (Windows et Mac ; sur Windows, aussi `Alt + 0201` sur le pavé numérique). Recopie les textes **exactement**,
accents et espaces compris : la vérification compare au caractère près. Dans les formules, tape toi-même les
signes du clavier (`-`, `*`, `/`).

### Étape 4 — Les outils du parcours

**Objectif** : Les paquets du parcours (jupyterlab, ipykernel, pandas, matplotlib, requests) sont installés **dans l'environnement du projet** `~/meteolocale/.venv`.

**À faire** :
1. Si ce n'est pas déjà fait (guide, étape 6) : dans Git Bash, `cd ~/meteolocale`, active l'environnement, puis
   `uv pip install -r 00-Parcours-Progressif/requirements.txt`.
2. Exécute la vérification : elle regarde ce qui est installé dans le dossier `~/meteolocale/.venv`.

**Résultat attendu** :
```
✅ OK
```

<details><summary>💡 Indice</summary>

Si `uv pip install` répond `File not found`, le parcours n'est pas au bon endroit : refais l'étape 6.1 du guide (`ls ~/meteolocale/00-Parcours-Progressif` doit montrer `requirements.txt`). 📖 Guide, étape 6.2.

</details>

````python
# Rien à écrire ici : fais-le dans Git Bash, puis exécute la vérification.
````

````python
# Vérification
from pathlib import Path
import os
PROJET = Path(os.environ.get("METEOLOCALE_PROJET", Path.home() / "meteolocale"))

venv = PROJET / ".venv"
assert (venv / "pyvenv.cfg").exists(), f"Pas d'environnement virtuel dans {PROJET} : refais l'étape 2."


def installe(nom):
    # Windows : .venv/Lib/site-packages ; macOS et Linux : .venv/lib/python3.13/site-packages
    return (venv / "Lib" / "site-packages" / nom).exists() or any(venv.glob(f"lib/python*/site-packages/{nom}"))


manquants = [nom for nom in ["jupyterlab", "ipykernel", "pandas", "matplotlib", "requests"] if not installe(nom)]
assert not manquants, f"Ces paquets ne sont pas installés dans {venv} : {manquants}. Dans Git Bash : cd ~/meteolocale, active l'environnement (le prompt doit montrer (meteolocale)), puis uv pip install -r 00-Parcours-Progressif/requirements.txt"
print("✅ OK")
````
