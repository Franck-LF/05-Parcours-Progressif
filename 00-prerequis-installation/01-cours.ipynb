## 1. Un notebook, c'est quoi ?

Un **notebook** (« carnet » en anglais) est un document qui mélange **du texte** et **des morceaux de code
qu'on peut exécuter**. Le fichier que tu lis en ce moment est un notebook : son nom finit par `.ipynb`.

Il est découpé en **cellules** : des cases posées les unes sous les autres, comme les lignes d'un tableau
Excel. Il en existe deux sortes :

| Sorte de cellule | Ce qu'elle contient | Comment la reconnaître |
|---|---|---|
| **Markdown** (texte) | Des explications, des titres, des tableaux | Du texte mis en forme, comme celui-ci |
| **Code** | Des instructions pour l'ordinateur | Un cadre gris, avec `[ ]` ou un petit triangle ▶ à gauche |

*Markdown* est le nom d'une façon simple d'écrire du texte mis en forme (titres avec `#`, gras avec `**`).
Tu n'as pas besoin de l'apprendre : tu vas surtout **lire** ces cellules.

## 2. Exécuter une cellule de code

dessous de la cellule.

Le code est écrit en **Python**, un langage : une façon d'écrire des instructions que l'ordinateur comprend. Tu
l'apprendras à partir du module 01. Aujourd'hui, on s'en sert comme d'une **calculatrice**.

Il compte les cellules exécutées depuis le début : la première a le numéro 1, la suivante 2, etc.

Pendant qu'une cellule travaille, tu vois `[*]` (ou un petit rond qui tourne) : attends qu'elle ait fini.

## 4. Le kernel : le moteur qui exécute et qui se souvient

Le **kernel** (« noyau » en anglais) est le **moteur** qui exécute le code des cellules. Il tourne en arrière-plan
tant que le notebook est ouvert. Son nom est affiché **en haut à droite** du notebook : chez toi, ce doit être
**`.venv`**, l'environnement virtuel du projet (voir le guide, étape 7.3).

Le kernel a une **mémoire** : il retient ce qu'on lui demande de retenir. Dans la cellule suivante, on lui
demande de retenir le nombre `25` sous le nom `temperature`. Le signe `=` veut dire « range la valeur de droite
sous le nom de gauche ». (C'est ce qu'on appellera une **variable** au module 01.)

```python
temperature = 25
```

🔮 **Prédis** : Cette cellule contient seulement le nom `temperature`. Qu'affiche-t-elle ? Écris ta réponse sur papier (ou dans ta tête), puis exécute.

```python
temperature
```

<details><summary>👀 Voir l'explication</summary>

`25`. Le kernel a **retenu** `temperature` à la cellule précédente, et il s'en souvient dans celle-ci. La mémoire
du kernel est partagée par **toutes** les cellules du notebook.

</details>

🔮 **Prédis** : Et cette cellule, qu'affiche-t-elle ? Écris ta réponse sur papier (ou dans ta tête), puis exécute.

```python
temperature = 30
```

<details><summary>👀 Voir l'explication</summary>

**Rien.** Ranger une valeur sous un nom n'affiche rien : le kernel retient `30`, en silence. Seul le numéro
`[n]` à gauche montre que la cellule a été exécutée.

</details>

🔮 **Prédis** : Avant d'exécuter : que va afficher cette cellule ? Écris ta réponse sur papier (ou dans ta tête), puis exécute.

```python
temperature
```

<details><summary>👀 Voir l'explication</summary>

`30`. La cellule précédente a **remplacé** `25` par `30` dans la mémoire du kernel. Le kernel retient
toujours **la dernière valeur** qu'on lui a donnée.

</details>

## 5. L'ordre d'exécution compte

Le kernel exécute les cellules **dans l'ordre où tu les lances**, pas dans l'ordre où elles sont écrites sur la
page. Les numéros `[n]` à gauche racontent l'histoire réelle.

Exemple : si tu remontes exécuter la cellule `temperature = 25` de la section 4, puis la cellule `temperature`
ci-dessous, tu verras `25`, même si la cellule `temperature = 30` est écrite plus bas sur la page.

🔮 **Prédis** : Avant d'exécuter : que va afficher cette cellule ? Écris ta réponse sur papier (ou dans ta tête), puis exécute.

```python
pluie = 0
pluie = pluie + 5
pluie
```

<details><summary>👀 Voir l'explication</summary>

`5`. Les lignes d'une cellule s'exécutent de haut en bas : d'abord `pluie` vaut `0`, puis on calcule `0 + 5` et on
range le résultat sous le même nom. La dernière ligne affiche `5`.

Si tu exécutes **encore** cette cellule, tu obtiens encore `5` : la première ligne remet `pluie` à `0` à chaque
fois.

</details>

**Conseil** : quand un résultat t'étonne, regarde les numéros `[n]`. Tu verras souvent qu'une cellule a été
oubliée ou exécutée deux fois.

## 6. ⚠️ Erreur fréquente : utiliser un nom que le kernel ne connaît pas

Que se passe-t-il si on demande au kernel un nom qu'on ne lui a **jamais** donné ? La cellule suivante plante
**volontairement**. Exécute-la quand même : c'est l'occasion d'apprendre à lire un message d'erreur.

```python
vent
```

**Lire le message, ligne par ligne** :

1. Va d'abord à **la dernière ligne** : c'est elle qui dit l'essentiel. Elle commence par le **type d'erreur** :
   `NameError` (erreur de nom).
2. Juste après : `name 'vent' is not defined` = « le nom `vent` n'est pas défini ». Le kernel ne connaît pas
   `vent` : on ne lui a jamais dit ce que c'était.
3. Au-dessus, une flèche `---->` ou un numéro de ligne montre **la ligne** de la cellule qui a posé problème.

**Les deux causes habituelles** :
- la cellule qui définit le nom **n'a pas été exécutée** (tu l'as sautée, ou le kernel a redémarré) ;
- le nom est **mal écrit** (`temperatur` au lieu de `temperature`) : Python ne devine jamais.

**La correction** : exécuter d'abord une cellule qui définit `vent` :

```python
vent = 11
vent
```

## 7. ⚠️ Erreur fréquente : du texte dans une cellule de code

Une cellule de **code** ne comprend que du Python. Si on y tape une phrase en français, le kernel ne sait pas la
lire. Exécute la cellule suivante (elle plante volontairement) :

```python
Il fait beau à Toulouse
```

**Lire le message** :

1. Dernière ligne : `SyntaxError: invalid syntax`. Une **SyntaxError** (erreur de syntaxe) veut dire « je ne
   comprends pas comment c'est écrit », comme une phrase sans verbe.
2. Au-dessus, la ligne fautive est recopiée, et un **chapeau `^`** pointe l'endroit où Python s'est perdu.

**Les corrections possibles** :
- si c'était une **explication**, la cellule devait être de type **Markdown** : clique sur la cellule, puis sur
  le mot `Python` en bas à droite de la cellule et choisis `Markdown` ;
- si tu voulais **afficher** cette phrase, il faut la mettre entre guillemets dans un `print` :

```python
print("Il fait beau à Toulouse")
```

## 8. Redémarrer le kernel

Quand le notebook se comporte bizarrement (des résultats qui ne collent plus, une cellule qui ne finit jamais),
la solution la plus simple est de **redémarrer le kernel** : on éteint le moteur et on le rallume.

Dans VS Code, en haut du notebook :
- **Restart** (flèche ronde ↻) : redémarre le kernel. **Toute sa mémoire est effacée** : `temperature`, `pluie`,
  `vent`… n'existent plus.
- **Run All** (▶▶) : exécute toutes les cellules, de la première à la dernière, dans l'ordre.

Après un **Restart**, si tu exécutes directement une cellule qui utilise `temperature`, tu obtiens…
une `NameError`, comme à la section 6 : le kernel a tout oublié. Il faut ré-exécuter les cellules depuis le haut.

> 💡 **Restart puis Run All** est le meilleur test pour savoir si ton notebook marche « pour de vrai » : si tout
> passe dans l'ordre, sans erreur, c'est bon.

**Essaie maintenant** : clique sur **Restart**, puis exécute **seulement** la cellule ci-dessous (sans
exécuter celles du dessus). Observe la `NameError`. Puis remonte exécuter la cellule `temperature = 25` et
reviens ici.

🔮 **Prédis** : Que va afficher cette cellule ? Écris ta réponse sur papier (ou dans ta tête), puis exécute.

```python
print(temperature)
```

## 10. Les gestes du terminal (pour les exercices)

Les exercices de ce module ne se font **pas** dans le notebook mais **dans Git Bash** (ou le Terminal sur Mac et
Linux). Le notebook sert seulement à **vérifier** ce que tu as fait : une cellule « Vérification » regarde ton
disque et affiche `✅ OK` si le dossier ou le fichier attendu est bien là.

**Organise ton écran** : VS Code d'un côté, Git Bash de l'autre. (Tu peux aussi utiliser le terminal intégré
de VS Code : menu **Terminal → New Terminal**.)

### 10.1 Se déplacer et créer

Tu les as déjà vues dans le guide. Une **option** est un réglage de la commande, écrit après un tiret (`-a`, `-p`).

| Commande | Ce qu'elle fait | Exemple |
|---|---|---|
| `pwd` | affiche le dossier où tu es | `pwd` → `/c/Users/sandrine/meteolocale` |
| `ls` | liste le contenu du dossier | `ls` |
| `ls -a` | liste **tout**, y compris les éléments cachés (nom qui commence par un point) | `ls -a` → `.  ..  .venv` |
| `cd nom` | entre dans le dossier `nom` | `cd bac-a-sable` |
| `cd ..` | remonte d'un cran | `cd ..` |
| `cd ~` | rentre dans ton dossier personnel, d'où que tu sois | `cd ~` |
| `mkdir nom` | crée un dossier `nom` **dans le dossier où tu es** | `mkdir essai-01` |
| `mkdir a b` | crée plusieurs dossiers d'un coup (séparés par un espace) | `mkdir lundi mardi` |
| `mkdir -p a/b/c` | crée `a`, puis `b` dans `a`, puis `c` dans `b` (et ne proteste pas s'ils existent) | `mkdir -p rapports/2026` |

**Le chemin d'un élément** : `rapports/2026` veut dire « le dossier `2026` qui est dans `rapports` ». Le `/`
sépare les étages. `~/meteolocale/bac-a-sable` part de ton dossier personnel, ce qui marche **d'où que tu sois**.

### 10.2 Créer et lire des fichiers texte

| Commande | Ce qu'elle fait |
|---|---|
| `touch notes.txt` | crée un fichier **vide** nommé `notes.txt` (*touch* = toucher : s'il n'existe pas, il est créé) |
| `echo "Bonjour"` | affiche `Bonjour` dans le terminal (*echo* = écho : il répète ce qu'on lui dit) |
| `echo "Bonjour" > bonjour.txt` | **écrit** `Bonjour` dans le fichier `bonjour.txt`, au lieu de l'afficher |
| `cat bonjour.txt` | affiche le contenu du fichier (*cat* vient de *concatenate*, mettre bout à bout) |

Le signe **`>`** (chevron) est un **aiguillage** : il envoie ce que la commande aurait affiché à l'écran **dans
un fichier**. Si le fichier existe déjà, il est **remplacé**. Ça marche avec n'importe quelle commande qui
affiche quelque chose : `pwd > ici.txt` écrit le chemin du dossier dans `ici.txt`, `ls > liste.txt` y écrit la
liste. C'est ainsi que les exercices vont « garder une preuve » de ce que le terminal a répondu.

### 10.3 L'environnement virtuel

| Commande | Ce qu'elle fait |
|---|---|
| `uv venv --python 3.13` | crée l'environnement virtuel `.venv` dans le dossier où tu es, avec Python 3.13 |
| `uv venv` | pareil, avec la version de Python que uv choisit (la plus récente installée) |
| `source .venv/Scripts/activate` | 🪟 **Windows (Git Bash)** : active l'environnement du dossier où tu es |
| `source .venv/bin/activate` | 🍎 🐧 **macOS / Linux** : active l'environnement du dossier où tu es |
| `which python` | affiche le chemin du Python utilisé (il doit contenir `.venv` si l'environnement est actif) |
| `python --version` | affiche la version de ce Python |
| `deactivate` | désactive l'environnement |
| `uv pip install requests` | installe le paquet `requests` dans l'environnement actif |
| `uv pip list` | liste les paquets installés dans l'environnement actif |

**Le témoin** : quand un environnement est actif, son nom apparaît entre parenthèses au début du prompt,
par exemple `(projet-a)`. Pas de parenthèses = pas d'environnement actif.

> ⚠️ Dans les exercices, **le dossier où tu es compte toujours**. Avant chaque commande, regarde le prompt ou
> tape `pwd`. La plupart des échecs viennent d'un `mkdir` ou d'un `uv venv` lancé au mauvais endroit.

### 10.4 Lire une erreur du terminal

Le terminal a aussi ses messages d'erreur. Les deux que tu verras le plus :

| Ce que tu tapes | Ce que le terminal répond | Comment le lire |
|---|---|---|
| `cd bac-a-sabel` | `bash: cd: bac-a-sabel: No such file or directory` | « aucun fichier ni dossier de ce nom » : faute de frappe, ou tu n'es pas au bon endroit. `ls` pour voir ce qui existe. |
| `mkdri essai` | `bash: mkdri: command not found` | « commande introuvable » : le **premier mot** est mal écrit (ou l'outil n'est pas installé). |

Le message reprend **ce que tu as tapé** (`bac-a-sabel`, `mkdri`) : compare-le lettre par lettre avec ce que tu
voulais taper.
