## 1. Pourquoi une liste ?

Au module 04, les 30 températures de septembre étaient collées dans une longue chaîne, et il fallait un tampon
pour les lire. Créer 30 variables (`t1`, `t2`, … `t30`) ne serait pas mieux.

Dans Excel, tu rangerais ces 30 valeurs dans **une colonne**. En Python, l'équivalent est une **liste** : une
seule variable qui contient plusieurs valeurs, **dans un ordre précis**.

````python
temperatures = [25.9, 23.2, 27.2, 26.1, 25.3]
print(temperatures)
print(type(temperatures))
````

- Une liste s'écrit entre **crochets** `[ ]`.
- Les valeurs, appelées **éléments**, sont séparées par des **virgules**.
- Son type est `list` (« liste » en anglais).

| Excel (colonne B) | Python |
|---|---|
| B1 : 25.9 | `temperatures[0]` |
| B2 : 23.2 | `temperatures[1]` |
| B3 : 27.2 | `temperatures[2]` |
| … | … |

Une liste peut contenir du texte, et même rien du tout :

````python
villes = ["Toulouse", "Montpellier", "Nîmes", "Perpignan", "Albi", "Tarbes"]
vide = []
print(villes)
print(vide)
````

`[]` est une **liste vide** : zéro élément. On s'en sert comme point de départ, comme le `total = 0` d'un
accumulateur.

Si tu as utilisé `.split()` au module 02, tu as déjà vu un résultat s'afficher entre crochets : c'était une liste.
On y revient en section 9.

`len()`, que tu connais pour les chaînes, donne aussi le **nombre d'éléments** d'une liste :

````python
print(len(villes))
````

💡 Rappel : `print` peut afficher **plusieurs valeurs d'un coup**, séparées par des virgules ; il met un espace
entre elles. On s'en servira pour comparer deux résultats sur une même ligne.

````python
print(len(villes), villes[0])
````

## 2. Lire un élément : l'index

Chaque élément a une **position**, appelée **index**. Comme pour les caractères d'une chaîne (module 02),
**on compte à partir de 0** : le premier élément est à l'index 0. Pense aux étages : le rez-de-chaussée est
l'étage 0.

On lit un élément en écrivant son index entre crochets **après** le nom de la liste :

````python
temperatures = [25.9, 23.2, 27.2, 26.1, 25.3]
print(temperatures[0])    # le 1er élément
print(temperatures[2])    # le 3e élément
````

| Index | 0 | 1 | 2 | 3 | 4 |
|---|---|---|---|---|---|
| Valeur | 25.9 | 23.2 | 27.2 | 26.1 | 25.3 |
| Index négatif | -5 | -4 | -3 | -2 | -1 |

Les **index négatifs** comptent depuis la fin : `-1` est le dernier, `-2` l'avant-dernier. Très pratique quand on
ne connaît pas la longueur de la liste.

````python
print(temperatures[-1])   # le dernier
print(temperatures[-2])   # l'avant-dernier
````

## 3. Prendre un morceau : la tranche

Comme pour les chaînes (module 02), une **tranche** `liste[debut:fin]` donne une **nouvelle liste** avec les
éléments de `debut` (compris) à `fin` (**non comprise**, comme dans `range`).

````python
semaine = [25.9, 23.2, 27.2, 26.1, 25.3, 22.7, 23.1]
print(semaine[0:3])    # index 0, 1, 2
print(semaine[:3])     # pareil : sans début, on part de 0
print(semaine[5:])     # sans fin, on va jusqu'au bout
print(semaine[-2:])    # les deux derniers
````

`[27.2, 26.1]` (index 2 et 3 ; le 4 n'est pas compris), puis `5` (index 1, 2, 3, 4, 5). Astuce : une tranche
`[a:b]` contient `b - a` éléments.

</details>

Une tranche ne modifie pas la liste de départ : elle en fabrique une copie partielle. Prendre une tranche trop
longue ne plante pas : `semaine[5:100]` donne simplement ce qui existe.

## 4. Modifier la liste : ajouter, insérer, retirer

Une liste peut grandir et rapetisser. On utilise des **méthodes** : des actions attachées à la valeur, qu'on
appelle avec un point, comme `.upper()` sur une chaîne au module 02.

### `.append(valeur)` : ajouter à la fin

C'est le geste le plus courant : ajouter une ligne en bas du tableau.

````python
releves = [25.9, 23.2]
releves.append(27.2)
print(releves)
releves.append(26.1)
print(releves)
````

### `.insert(position, valeur)` : insérer ailleurs

On donne d'abord la position, puis la valeur. Les éléments suivants se décalent, comme quand on insère une ligne
dans Excel.

````python
villes = ["Toulouse", "Nîmes", "Tarbes"]
villes.insert(1, "Montpellier")     # devient l'élément d'index 1
print(villes)
````

### `.remove(valeur)` : retirer une valeur

On donne la **valeur** à retirer (pas sa position). Si elle apparaît plusieurs fois, seule la **première** est
retirée.

````python
villes = ["Toulouse", "Paris", "Albi"]
villes.remove("Paris")              # Paris n'est pas en Occitanie !
print(villes)
````

### `.pop()` : retirer et récupérer

`.pop()` retire le **dernier** élément **et le donne** : on peut le ranger dans une variable. `.pop(0)` retire le
premier.

````python
releves = [25.9, 23.2, 27.2, 27.2]
doublon = releves.pop()             # le dernier relevé était en double
print(doublon)
print(releves)
````

🔮 **Prédis** : Avant d'exécuter : que va afficher cette cellule ? Écris ta réponse sur papier (ou dans ta tête), puis exécute.

````python
liste = [1, 2, 3]
liste.append(4)
liste.insert(0, 0)
liste.remove(2)
dernier = liste.pop()
print(liste)
print(dernier)
````

<details><summary>👀 Voir l'explication</summary>

On suit pas à pas : `[1, 2, 3]` → append 4 → `[1, 2, 3, 4]` → insert(0, 0) → `[0, 1, 2, 3, 4]` → remove(2)
→ `[0, 1, 3, 4]` → pop() retire 4. Affichage : `[0, 1, 3]` puis `4`.

</details>

## 6. Parcourir une liste avec `for`

Au module 04, `for` parcourait une chaîne lettre par lettre. Sur une liste, il la parcourt **élément par
élément** : « pour chaque ligne de la colonne ». Tous les gestes du module 04 (accumulateur, compteur, record)
marchent à l'identique.

````python
pluies = [17.4, 0.7, 0.0, 10.4, 0.0]
total = 0
nb_jours_pluie = 0
for pluie in pluies:
    total = total + pluie
    if pluie >= 1:
        nb_jours_pluie = nb_jours_pluie + 1
print(f"Total : {total:.1f} mm, {nb_jours_pluie} jours de pluie")
````

### Construire une nouvelle liste dans une boucle

Le geste clé de ce module : partir d'une liste **vide** et y **ajouter** (`append`) les éléments qui nous
intéressent. C'est l'équivalent d'un **filtre** Excel.

````python
temperatures = [25.9, 23.2, 27.2, 26.1, 22.7]
chaudes = []                       # 1. une liste vide avant la boucle
for t in temperatures:
    if t > 25:
        chaudes.append(t)          # 2. on ajoute ce qui passe le filtre
print(chaudes)                     # 3. on utilise le résultat après
````

C'est exactement l'accumulateur du module 04, avec `[]` au lieu de `0` et `.append()` au lieu de `+`.

On peut aussi **transformer** chaque élément :

````python
temperatures = [25.9, 23.2, 27.2]
en_fahrenheit = []
for t in temperatures:
    en_fahrenheit.append(round(t * 9 / 5 + 32, 1))
print(en_fahrenheit)
````

### `enumerate` : le numéro et la valeur

Souvent, on veut aussi le **numéro** de l'élément (« jour 3 »). `enumerate(liste)` donne, à chaque tour, **deux
choses** : le numéro et la valeur. On les range dans **deux variables**, séparées par une virgule.

````python
temperatures = [25.9, 23.2, 27.2]
for numero, t in enumerate(temperatures):
    print(f"index {numero} : {t} °C")
````

| Tour | ce que donne `enumerate` | `numero` | `t` |
|---|---|---|---|
| 1 | `0, 25.9` | 0 | 25.9 |
| 2 | `1, 23.2` | 1 | 23.2 |
| 3 | `2, 27.2` | 2 | 27.2 |

Par défaut, la numérotation commence à 0 (comme les index). Pour les jours du mois, on veut commencer à 1 :
on donne le départ en deuxième valeur, `enumerate(temperatures, 1)`.

````python
for jour, t in enumerate(temperatures, 1):
    print(f"Jour {jour} : {t} °C")
````

### Parcourir deux listes en même temps avec les index

Quand deux listes vont ensemble (même ordre : la ville d'index 0 a la température d'index 0), on parcourt les
**index** avec `range(len(...))`, puis on lit chaque liste à cet index :

````python
villes = ["Toulouse", "Montpellier", "Nîmes"]
maximales = [25.9, 29.2, 29.2]
for i in range(len(villes)):
    print(f"{villes[i]} : {maximales[i]} °C")
````

`len(villes)` vaut 3, donc `range(3)` donne 0, 1, 2 : exactement les index de la liste.

### Lire l'élément d'avant : `liste[i - 1]`

Entre les crochets, on peut mettre un **calcul**. Si `i` est l'index du jour, `i - 1` est l'index de la **veille**.
Pour comparer chaque jour à la veille, on commence à `i = 1` : le premier jour (index 0) n'a pas de veille.

````python
temperatures = [25.9, 23.2, 27.2, 26.1]
for i in range(1, len(temperatures)):
    aujourdhui = temperatures[i]
    veille = temperatures[i - 1]
    print(f"index {i} : {veille} -> {aujourdhui}, variation {round(aujourdhui - veille, 1)}")
````

| `i` | `i - 1` | `temperatures[i - 1]` (veille) | `temperatures[i]` (jour) |
|---|---|---|---|
| 1 | 0 | 25.9 | 23.2 |
| 2 | 1 | 23.2 | 27.2 |
| 3 | 2 | 27.2 | 26.1 |

🔮 **Prédis** : Avant d'exécuter : que va afficher cette cellule ? Écris ta réponse sur papier (ou dans ta tête), puis exécute.

````python
pluies = [0.0, 3.5, 3.5, 1.2]
for i in range(1, len(pluies)):
    if pluies[i] == pluies[i - 1]:
        print(f"Jour {i + 1} : comme la veille")
````

`range(1, ...)`, on ne lit jamais `pluies[-1]` par erreur au premier tour.

</details>

## 7. Résumer une liste : `len`, `sum`, `min`, `max`

Python fournit les mêmes outils que les formules d'Excel :

| Excel | Python | Résultat |
|---|---|---|
| `NB(B1:B30)` | `len(liste)` | nombre d'éléments |
| `SOMME(B1:B30)` | `sum(liste)` | total |
| `MIN(B1:B30)` | `min(liste)` | plus petite valeur |
| `MAX(B1:B30)` | `max(liste)` | plus grande valeur |
| `MOYENNE(B1:B30)` | `sum(liste) / len(liste)` | moyenne |

````python
temperatures = [25.9, 23.2, 27.2, 26.1, 25.3]
print(len(temperatures))
print(sum(temperatures))
print(min(temperatures))
print(max(temperatures))
moyenne = sum(temperatures) / len(temperatures)
print(f"Moyenne : {moyenne:.1f} °C")
````

Ces outils font **en une ligne** ce que tes boucles du module 04 faisaient à la main : `sum` est un
accumulateur, `max` un record. Tu sais donc ce qu'ils font à l'intérieur.

`sum(temperatures)` peut afficher un nombre comme `127.69999999999999` au lieu de `127.7` : c'est une petite
imprécision normale des nombres à virgule dans un ordinateur. On arrondit à l'affichage avec `round` ou `:.1f`.

🔮 **Prédis** : Avant d'exécuter : que va afficher cette cellule ? Écris ta réponse sur papier (ou dans ta tête), puis exécute.

````python
pluies = [3, 0, 5, 2]
print(max(pluies) - min(pluies))
print(sum(pluies) / len(pluies))
````

<details><summary>👀 Voir l'explication</summary>

`5` (5 - 0) puis `2.5` (10 / 4). Une division `/` donne toujours un `float`.

</details>

`min` et `max` marchent aussi sur des textes : ils comparent dans l'ordre alphabétique (`min(villes)` donne la
première ville dans l'ordre alphabétique).

## 8. Trier : `sorted()` et `.sort()`

Deux façons de trier, comme le bouton « Trier de A à Z » d'Excel :

- `sorted(liste)` **fabrique une nouvelle liste** triée ; l'originale ne bouge pas.
- `liste.sort()` **trie la liste elle-même** (et ne renvoie rien, comme `.append()`).

````python
temperatures = [25.9, 23.2, 27.2, 26.1]
triees = sorted(temperatures)
print(triees)
print(temperatures)       # inchangée

temperatures.sort()
print(temperatures)       # triée, cette fois
````

Le tri est **croissant** : du plus petit au plus grand (ou de A à Z pour du texte). Pour avoir les **3 plus
grandes** valeurs, on trie puis on prend les 3 dernières avec une tranche :

````python
temperatures = [25.9, 23.2, 27.2, 26.1, 25.3, 22.7, 27.1]
podium = sorted(temperatures)[-3:]
print(podium)
````

🔮 **Prédis** : Avant d'exécuter : que va afficher cette cellule ? Écris ta réponse sur papier (ou dans ta tête), puis exécute.

````python
villes = ["Tarbes", "Albi", "Nîmes"]
resultat = villes.sort()
print(resultat)
print(villes)
````

<details><summary>👀 Voir l'explication</summary>

`None` puis `['Albi', 'Nîmes', 'Tarbes']`. `.sort()` trie `villes` sur place et ne renvoie **rien** (`None`) :
`resultat` ne contient donc pas la liste triée. Si tu veux une liste triée dans une nouvelle variable, utilise
`sorted(villes)`.

</details>

## 9. Du texte à la liste : `.split()`

Le capteur envoie toujours ses mesures dans une chaîne : `"12.0;13.5;11.8"`. Au module 04, il fallait un tampon.
Maintenant, `.split(";")` découpe la chaîne à chaque `;` et donne une **liste de morceaux** :

````python
mesures = "12.0;13.5;11.8"
morceaux = mesures.split(";")
print(morceaux)
print(type(morceaux[0]))
````

Attention : les morceaux sont des **textes** (`'12.0'`, avec apostrophes). Pour calculer, on fabrique une liste de
nombres avec une boucle et `float()` :

````python
valeurs = []
for morceau in morceaux:
    valeurs.append(float(morceau))
print(valeurs)
print(sum(valeurs))
````

````python
ligne = "2026-09-01,Toulouse,12.0,25.9"
morceaux = ligne.split(",")
print(len(morceaux))
print(morceaux[1])
print(float(morceaux[3]) - float(morceaux[2]))
````

## 10. La compréhension de liste : la boucle en une ligne

Tu as écrit plusieurs fois ce motif : liste vide, boucle, `append`. Python propose une écriture plus courte, la
**compréhension de liste**. Elle ne fait **rien de nouveau** : c'est la même boucle, écrite autrement.

Version boucle (que tu connais) :

````python
morceaux = ["12.0", "13.5", "11.8"]
valeurs = []
for m in morceaux:
    valeurs.append(float(m))
print(valeurs)
````

Version compréhension :

````python
valeurs = [float(m) for m in morceaux]
print(valeurs)
````

Lis-la de gauche à droite : « une liste `[ ]` de `float(m)`, **pour chaque** `m` **dans** `morceaux` ».

| Boucle | Compréhension |
|---|---|
| `valeurs = []` | les crochets `[ ... ]` |
| `for m in morceaux:` | `for m in morceaux` (sans les deux-points) |
| `valeurs.append(float(m))` | `float(m)` (écrit **en premier**) |

Dans Excel, c'est comme écrire une formule en haut d'une nouvelle colonne et la tirer jusqu'en bas.

### Avec un filtre

On peut ajouter un `if` à la fin : seuls les éléments qui passent le test sont gardés.

````python
temperatures = [25.9, 23.2, 27.2, 26.1, 22.7]
chaudes = [t for t in temperatures if t > 25]
print(chaudes)
````

C'est exactement la boucle de la section 6 (liste vide, `for`, `if`, `append`), en une ligne.

🔮 **Prédis** : Avant d'exécuter : que va afficher cette cellule ? Écris ta réponse sur papier (ou dans ta tête), puis exécute.

````python
jours = [1, 2, 3, 4, 5, 6]
pairs = [j * 10 for j in jours if j % 2 == 0]
print(pairs)
````

<details><summary>👀 Voir l'explication</summary>

`[20, 40, 60]`. On garde les jours pairs (2, 4, 6) et on écrit `j * 10` pour chacun.

</details>

💡 **Conseil** : si la compréhension te paraît obscure, écris d'abord la boucle. Les deux sont justes. La
compréhension devient naturelle avec l'habitude ; tu la retrouveras souvent en data.
