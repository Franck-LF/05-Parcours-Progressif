## 1. Afficher avec `print`

En Python, un nom suivi de parenthèses, comme `print(...)`, s'appelle une **fonction** : un outil prêt à l'emploi
qu'on « appelle » en lui donnant quelque chose entre les parenthèses. `print` est la première que tu rencontres.

````python
print("Bonjour MétéoLocale")
print(25)
````

## 3. Les variables : des étiquettes sur des boîtes

Une **variable** est un **nom** qui désigne une **valeur**. Imagine une boîte de rangement avec une étiquette
collée dessus : l'étiquette, c'est le nom ; ce qu'il y a dans la boîte, c'est la valeur.

On crée une variable avec le signe **`=`** :

````python
ville = "Toulouse"
temp_max = 25.9
print(ville, temp_max)
````

- `ville = "Toulouse"` se lit : **« range `"Toulouse"` dans la boîte `ville` »**. Ce n'est **pas** une égalité de
  mathématiques : `=` veut dire « ranger » (on dit aussi **affecter** une valeur).
- À gauche du `=` : toujours le **nom**. À droite : la **valeur** (ou un calcul).
- Ensuite, partout où tu écris `ville` (sans guillemets), Python va chercher ce qu'il y a dans la boîte.
- Créer une variable n'affiche rien : il faut un `print` pour voir son contenu.

🔮 **Prédis** : Avant d'exécuter : que va afficher cette cellule ? Écris ta réponse sur papier (ou dans ta tête), puis exécute.

````python
ville = "Toulouse"
print("ville")
print(ville)
````

<details><summary>👀 Voir l'explication</summary>

```
ville
Toulouse
```
Avec des guillemets, `"ville"` est un **texte** : il s'affiche tel quel. Sans guillemets, `ville` est le **nom
d'une variable** : Python affiche ce qu'elle contient.

</details>

### Changer la valeur d'une variable

On peut **remplacer** le contenu d'une boîte : il suffit de ranger autre chose sous le même nom. L'ancienne
valeur est perdue.

🔮 **Prédis** : Avant d'exécuter : que va afficher cette cellule ? Écris ta réponse sur papier (ou dans ta tête), puis exécute.

````python
temp_max = 25.9
temp_max = 27.2
print(temp_max)
````

<details><summary>👀 Voir l'explication</summary>

`27.2`. La deuxième ligne remplace `25.9` par `27.2`. Une variable contient **une seule valeur à la fois** : la
dernière rangée.

</details>

### Choisir un bon nom

| Règle | ✅ Correct | ❌ Interdit ou déconseillé |
|---|---|---|
| Lettres minuscules, chiffres et `_` (tiret bas) | `temp_max`, `jour2` | `temp-max` (le `-` est une soustraction) |
| Pas d'espace : on le remplace par `_` | `pluie_mm` | `pluie mm` |
| Ne commence pas par un chiffre | `jour1` | `1jour` |
| Majuscules et minuscules sont différentes | `ville` et `Ville` sont **deux** variables | — |
| Un nom qui dit ce qu'il contient | `temp_min` | `x`, `truc` |
| Dans les **noms** de variables, évite les accents (autorisés, mais source d'erreurs). Dans les **textes** entre guillemets, les accents sont normaux : `"Écart"` | `temperature` | `température` |

## 4. Les types : quatre sortes de valeurs

Chaque valeur a un **type** : sa catégorie. Comme dans Excel, où une cellule contient un nombre, un texte ou une
date, et où Excel ne les traite pas pareil. Les quatre types de base :

| Type | Nom complet | Ce que c'est | Exemples |
|---|---|---|---|
| `int` | *integer*, entier | un nombre **sans** virgule | `11`, `30`, `-3` |
| `float` | nombre à virgule (*flottante*) | un nombre **avec** un point décimal | `25.9`, `17.4`, `12.0` |
| `str` | *string*, chaîne de caractères | du **texte**, entre guillemets | `"Toulouse"`, `"25.9"` |
| `bool` | booléen | **vrai** ou **faux** : `True` ou `False` | `True`, `False` |

La fonction **`type(...)`** dit de quel type est une valeur :

````python
print(type(11))
print(type(25.9))
print(type("Toulouse"))
print(type(True))
````

`<class 'int'>` se lit simplement « type `int` ». (Le mot *class* sera expliqué au module 11 : ignore-le.)

🔮 **Prédis** : Avant d'exécuter : que va afficher cette cellule ? Écris ta réponse sur papier (ou dans ta tête), puis exécute.

````python
print(type("25.9"))
````

<details><summary>👀 Voir l'explication</summary>

`<class 'str'>`. Les guillemets font de `"25.9"` un **texte**, même s'il ressemble à un nombre. On ne peut pas
calculer avec (on verra au module 02 comment le transformer en nombre).

</details>

🔮 **Prédis** : Avant d'exécuter : que va afficher cette cellule ? Écris ta réponse sur papier (ou dans ta tête), puis exécute.

````python
print(type(12.0))
````

<details><summary>👀 Voir l'explication</summary>

`<class 'float'>`. Le point décimal suffit : `12.0` est un `float`, même si la partie après la virgule vaut zéro.
`12` (sans point) serait un `int`.

</details>

**Le type `bool`** sert à répondre à une question par oui ou non : « a-t-il plu ? ». Il n'a que deux valeurs,
`True` (vrai) et `False` (faux), **avec une majuscule**. On s'en servira beaucoup au module 03.

````python
il_a_plu = True
print(il_a_plu, type(il_a_plu))
````

### ⚠️ Erreur fréquente : la virgule décimale

En France, on écrit `25,9`. **En Python, le séparateur décimal est le point** : `25.9`. Avec une virgule, Python
ne plante pas, mais il comprend **autre chose** :

````python
temp_max = 25,9
print(temp_max)
print(type(temp_max))
````

Python affiche `(25, 9)` : il a vu **deux nombres**, `25` et `9`, séparés par une virgule, et les a regroupés en un
paquet (le type `tuple`, qu'on verra au module 06). Ce n'est pas un `float`, et les calculs vont planter ou
donner n'importe quoi plus loin.

**Correction** : toujours le point → `temp_max = 25.9`. Réflexe : en cas de doute, `type(...)` doit afficher
`float`.

## 5. Calculer : les opérations

| Opération | Symbole | Exemple | Résultat |
|---|---|---|---|
| Addition | `+` | `12 + 8` | `20` |
| Soustraction | `-` | `25 - 12` | `13` |
| Multiplication | `*` (une étoile) | `4.5 * 3` | `13.5` |
| Division | `/` | `38 / 2` | `19.0` |
| Division entière | `//` | `30 // 7` | `4` |
| Reste de la division | `%` | `30 % 7` | `2` |
| Puissance | `**` (deux étoiles) | `2 ** 3` | `8` |

Les espaces autour des symboles ne changent rien au calcul, mais rendent le code plus lisible : écris
`12 + 8` plutôt que `12+8`.

🔮 **Prédis** : Avant d'exécuter : que va afficher cette cellule ? Écris ta réponse sur papier (ou dans ta tête), puis exécute.

<details><summary>👀 Voir l'explication</summary>

`5.0`, et non `5`. **La division `/` donne toujours un `float`**, même quand le résultat tombe juste.

</details>

### Division entière `//` et reste `%`

Septembre a 30 jours. Combien de **semaines complètes**, et combien de **jours restants** ?

- `30 // 7` : combien de fois 7 tient dans 30, **sans les morceaux** → `4` semaines complètes.
- `30 % 7` : ce qu'il **reste** après avoir fait ces 4 paquets de 7 (`4 * 7 = 28`, et `30 - 28 = 2`) → `2` jours.

C'est la division « de l'école primaire » : 30 divisé par 7 = 4, reste 2.

````python
semaines = 30 // 7
jours_restants = 30 % 7
print(semaines, "semaines et", jours_restants, "jours")
````

5 tient 3 fois dans 17 (`3 * 5 = 15`), et il reste `17 - 15 = 2`.

</details>

## 6. La priorité des opérations

Comme en mathématiques : **`**` d'abord, puis `*` `/` `//` `%`, puis `+` `-`**. Les **parenthèses** passent avant
tout. Les calculs de même niveau se font de gauche à droite.

🔮 **Prédis** : Avant d'exécuter : que va afficher cette cellule ? Écris ta réponse sur papier (ou dans ta tête), puis exécute.

````python
print(12 + 26 / 2)
print((12 + 26) / 2)
````

<details><summary>👀 Voir l'explication</summary>

```
25.0
19.0
```
Sans parenthèses, la division passe **avant** l'addition : `26 / 2 = 13.0`, puis `12 + 13.0 = 25.0`. Ce n'est
pas la moyenne de 12 et 26 !

Avec parenthèses, l'addition est faite d'abord : `38 / 2 = 19.0`. **Pour une moyenne, toujours des parenthèses
autour de la somme.**

</details>

## 7. Calculer avec des variables

Un calcul peut utiliser des variables, et son résultat peut être rangé dans une nouvelle variable. Python
calcule d'abord **à droite** du `=`, puis range le résultat **à gauche**.

````python
temp_min = 12
temp_max = 26
ecart = temp_max - temp_min
moyenne = (temp_min + temp_max) / 2
print("Écart :", ecart)
print("Moyenne :", moyenne)
````

### Mettre à jour une variable à partir d'elle-même

Pour **ajouter** quelque chose au contenu d'une boîte, on écrit `total = total + ...`. Python lit d'abord la
droite (l'ancienne valeur de `total`, plus le nouveau morceau), puis range le résultat dans `total`.

🔮 **Prédis** : Avant d'exécuter : que va afficher cette cellule ? Écris ta réponse sur papier (ou dans ta tête), puis exécute.

````python
pluie_totale = 0
pluie_totale = pluie_totale + 4
pluie_totale = pluie_totale + 6
print(pluie_totale)
````

`10`. Au départ `0` ; puis `0 + 4 = 4` ; puis `4 + 6 = 10`. C'est un **cumul**, comme une colonne « Total » qu'on
met à jour ligne après ligne dans Excel.

</details>

## 8. Les nombres à virgule et `round`

Si un calcul mélange un `int` et un `float`, le résultat est un `float` : `3 + 2.0` donne `5.0`.

Les `float` ont un défaut : l'ordinateur les stocke avec une toute petite imprécision (il compte en binaire, pas
en décimal). D'habitude elle est invisible, mais parfois elle saute aux yeux :

````python
ecart = 25.9 - 12.0
print(ecart)
````

On attendait `13.9`, Python affiche `13.899999999999999`. Ce n'est **pas** une erreur de ton code : c'est normal
pour tous les langages. La solution : **arrondir** avec la fonction `round(valeur, nombre_de_chiffres)`.

````python
ecart = round(25.9 - 12.0, 1)
print(ecart)
````

🔮 **Prédis** : Avant d'exécuter : que va afficher cette cellule ? Écris ta réponse sur papier (ou dans ta tête), puis exécute.

````python
print(round(18.6666, 1))
print(round(18.6666, 2))
print(round(18.6666))
````

<details><summary>👀 Voir l'explication</summary>

```
18.7
18.67
19
```
Le deuxième nombre dans `round(..., 1)` dit **combien de chiffres après la virgule** garder. Sans ce deuxième
nombre, `round` arrondit à l'**entier** le plus proche, et donne un `int`.

</details>

## 9. Tout ensemble : convertir des °C en °F

Les clients américains de MétéoLocale veulent les températures en **degrés Fahrenheit**. La formule :
`°F = °C * 9 / 5 + 32`. En Python : `temp_f = temp_c * 9 / 5 + 32`.

````python
temp_c = 25.9                      # température en degrés Celsius
temp_f = temp_c * 9 / 5 + 32       # conversion en degrés Fahrenheit
temp_f = round(temp_f, 1)          # un chiffre après la virgule suffit
print(temp_c, "°C =", temp_f, "°F")
````

Ici, pas besoin de parenthèses : `*` et `/` passent avant `+`, donc Python calcule `25.9 * 9 / 5` d'abord, puis
ajoute `32`. C'est exactement la formule.
