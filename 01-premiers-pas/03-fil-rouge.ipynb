## Comment travailler

Comme dans les exercices : remplace les `...`, exécute ta cellule (`Maj + Entrée`), puis la cellule
**Vérification** (`✅ OK` = réussi).

**Si la vérification échoue** : un long message apparaît, surtout en anglais, qui se termine par une ligne comme
`AssertionError: ville vaut 'albi' au lieu de 'Albi'…`. `AssertionError` veut seulement dire **« la vérification a
échoué »**. Lis **uniquement le texte en français après les deux-points** : il te dit quoi corriger. Tout le reste
(lignes en anglais, numéros, flèches, lignes `assert …` de la cellule) peut être ignoré. Corrige ta cellule,
ré-exécute-la, puis ré-exécute la vérification.

**Caractères spéciaux** : les signes difficiles à taper (`°`, `É`, les pictogrammes comme ⛈ 🌧) se
**copient-collent depuis l'énoncé** : sélectionne-les à la souris, `Ctrl + C`, puis `Ctrl + V` dans ta cellule
(`Cmd` au lieu de `Ctrl` sur Mac). Au clavier français : `°` = `Maj + )` (Windows et Mac) ; `É` = `Verr. Maj` puis
`é` (Windows et Mac ; sur Windows, aussi `Alt + 0201` sur le pavé numérique). Recopie les textes **exactement**,
accents et espaces compris : la vérification compare au caractère près. Dans les formules, tape toi-même les
signes du clavier (`-`, `*`, `/`).
