## 1. Pourquoi un dictionnaire ?

Une ligne du fichier MétéoLocale, découpée avec `.split(",")` (module 05), donne une liste :

```python
releve_liste = ["2026-09-01", "Toulouse", 12.0, 25.9, 17.4, 11]
print(releve_liste[3])
```

Ça marche… mais que veut dire `releve_liste[3]` ? Il faut se souvenir que la position 3 est la température
maximale. Et dans six mois, plus personne ne s'en souviendra.

Dans Excel, chaque colonne a un **en-tête** : `date`, `ville`, `temp_max`… On lit une valeur par le nom de sa
colonne, pas par son numéro. En Python, l'outil qui fait ça s'appelle un **dictionnaire** : une **fiche avec des
rubriques**, comme une fiche client avec « Nom : … », « Ville : … ».

```python
releve = {"date": "2026-09-01", "ville": "Toulouse", "temp_max": 25.9, "pluie_mm": 17.4}
print(releve)
print(type(releve))
```

- Un dictionnaire s'écrit entre **accolades** `{ }`.
- Il contient des **paires** `clé: valeur`, séparées par des virgules.
- La **clé** est le nom de la rubrique (ici un texte : `"ville"`). La **valeur** est son contenu (`"Toulouse"`).
- Entre la clé et la valeur : deux-points `:`.
- Son type est `dict` (abréviation de *dictionary*).

Pourquoi « dictionnaire » ? Dans un vrai dictionnaire, on cherche un **mot** (la clé) pour trouver sa
**définition** (la valeur).

| Clé (l'en-tête) | Valeur (le contenu) |
|---|---|
| `"date"` | `"2026-09-01"` |
| `"ville"` | `"Toulouse"` |
| `"temp_max"` | `25.9` |
| `"pluie_mm"` | `17.4` |

`len()` donne le nombre de **paires** (de rubriques) :

```python
print(len(releve))
```

Pour la lisibilité, on écrit souvent un dictionnaire sur plusieurs lignes. Python l'accepte tant que les
accolades ne sont pas fermées :

```python
releve = {
    "date": "2026-09-01",
    "ville": "Toulouse",
    "temp_max": 25.9,
    "pluie_mm": 17.4,
}
print(releve)
```

## 3. Modifier, ajouter, supprimer

### Modifier et ajouter : la même écriture

`releve["cle"] = valeur` **remplace** la valeur si la clé existe, et **ajoute** une nouvelle rubrique sinon.
C'est comme retaper une cellule Excel, ou ajouter une colonne.

```python
releve = {"date": "2026-09-01", "ville": "Toulouse", "temp_max": 52.9}
releve["temp_max"] = 25.9        # la clé existe : on CORRIGE
releve["vent_kmh"] = 11          # la clé n'existe pas : on AJOUTE
print(releve)
```

Une clé est **unique** dans un dictionnaire : il ne peut pas y avoir deux rubriques `"temp_max"`.

### Supprimer : `del` ou `.pop()`

`del` (pour *delete*, « supprimer ») retire une rubrique. `.pop("cle")` la retire **et donne sa valeur**, comme
pour les listes.

```python
releve = {"ville": "Toulouse", "temp_max": 25.9, "commentaire": "test capteur"}
del releve["commentaire"]
print(releve)

temperature = releve.pop("temp_max")
print(temperature)
print(releve)
```

## 4. Une rubrique peut manquer : `in` et `.get()`

Certains capteurs ne mesurent pas le vent. Avant de lire une clé, on peut vérifier qu'elle existe avec `in`
(comme pour les listes, mais `in` regarde **les clés**) :

```python
releve = {"ville": "Albi", "temp_max": 26.8}
print("temp_max" in releve)
print("vent_kmh" in releve)
print("Albi" in releve)          # False : "Albi" est une VALEUR, pas une clé
```

Plus court : `.get(cle)` renvoie la valeur si la clé existe, et `None` (« rien », module 05) sinon, **sans
planter**. On peut donner une **valeur par défaut** en deuxième : `.get(cle, 0)` renvoie 0 si la clé manque.

```python
print(releve.get("temp_max"))
print(releve.get("vent_kmh"))
print(releve.get("vent_kmh", 0))
```

## 5. Les tuples et le déballage

Avant de parcourir un dictionnaire, il nous faut un dernier outil.

### Le tuple : une petite liste figée

Un **tuple** ressemble à une liste, mais s'écrit avec des **parenthèses** et **ne peut pas être modifié**. On s'en
sert pour regrouper quelques valeurs qui vont ensemble, comme les coordonnées d'une station (latitude,
longitude).

```python
station_toulouse = (43.6, 1.44)
print(station_toulouse)
print(type(station_toulouse))
print(station_toulouse[0])       # on lit par index, comme une liste
print(len(station_toulouse))
```

### ⚠️ Erreur fréquente : modifier un tuple

```python
station_toulouse = (43.6, 1.44)
station_toulouse[0] = 43.7
```

1. Dernière ligne : `TypeError: 'tuple' object does not support item assignment`. Traduction : « un tuple
   n'accepte pas qu'on remplace un de ses éléments ».
2. C'est voulu : un tuple est **figé**. C'est une garantie que personne ne le modifiera par erreur.

✅ Correction : si tu dois modifier les éléments, utilise une liste `[43.6, 1.44]` ; sinon, crée un nouveau tuple
`station_toulouse = (43.7, 1.44)`.

### Le déballage : une variable par élément

On peut **déballer** un tuple : ranger chacun de ses éléments dans sa propre variable, en une ligne. Il faut
**autant de variables que d'éléments**.

```python
station_toulouse = (43.6, 1.44)
latitude, longitude = station_toulouse
print(f"Latitude {latitude}, longitude {longitude}")
```

`sorted` sait trier une liste de tuples : il compare d'abord le **premier élément** de chaque tuple, et ne

## 6. Parcourir un dictionnaire

Un dictionnaire donne trois « vues » :

| Méthode | Ce qu'elle donne | Excel |
|---|---|---|
| `.keys()` | les clés | les en-têtes de colonnes |
| `.values()` | les valeurs | le contenu des cellules |
| `.items()` | les paires `(clé, valeur)`, sous forme de tuples | les deux |

```python
pluie_par_ville = {"Toulouse": 17.4, "Montpellier": 14.0, "Nîmes": 10.0, "Tarbes": 1.6}
print(pluie_par_ville.keys())
print(pluie_par_ville.values())
print(pluie_par_ville.items())
```

Ce ne sont pas tout à fait des listes : ce sont des **vues** sur le dictionnaire, qu'on peut parcourir avec `for`
mais pas indexer (`pluie_par_ville.keys()[0]` plante). Quand on veut un index ou une tranche, on les convertit
en vraie liste avec `list()`, comme on convertit avec `int()` ou `str()` :

```python
villes = list(pluie_par_ville.keys())
print(villes)
print(villes[0])
print(sum(pluie_par_ville.values()))     # sum marche directement sur les valeurs
```

### La boucle `for` sur un dictionnaire

- `for cle in d:` parcourt **les clés** (c'est pareil que `for cle in d.keys():`).
- `for cle, valeur in d.items():` donne à chaque tour un tuple `(clé, valeur)`, **déballé** dans deux variables.
  C'est la forme la plus utilisée.

```python
for ville in pluie_par_ville:
    print(ville)
```

```python
for ville, pluie in pluie_par_ville.items():
    print(f"{ville} : {pluie} mm")
```

Le déballage marche aussi dans une compréhension de liste (module 05) : on écrit les deux variables après `for`.

```python
villes_arrosees = [ville for ville, pluie in pluie_par_ville.items() if pluie >= 10]
print(villes_arrosees)
```

🔮 **Prédis** : Avant d'exécuter : que va afficher cette cellule ? Écris ta réponse sur papier (ou dans ta tête), puis exécute.

```python
vents = {"Toulouse": 11, "Tarbes": 52, "Nîmes": 42}
nb_forts = 0
for ville, vent in vents.items():
    if vent >= 40:
        nb_forts = nb_forts + 1
        print(ville)
print(nb_forts)
```

<details><summary>👀 Voir l'explication</summary>

`Tarbes`, `Nîmes`, puis `2`. Un compteur (module 04), dans une boucle sur les paires du dictionnaire. Un
dictionnaire garde l'**ordre d'ajout** des clés : Tarbes est lue avant Nîmes.

</details>

### Classer : fabriquer une liste de tuples

On peut **fabriquer** des tuples à partir de variables, puis les ranger dans une liste avec `append`. Exemple :
classer des villes d'après leur moyenne. On met la moyenne **en premier** dans chaque tuple, pour que `sorted`
trie d'après elle (section 5, « Trier des tuples »).

```python
moyennes = {"Toulouse": 24.55, "Albi": 26.9, "Tarbes": 22.55}
paires = []
for ville, moyenne in moyennes.items():
    paires.append((moyenne, ville))
print(paires)
```

Attention aux **doubles parenthèses** dans `paires.append((moyenne, ville))` :

- les parenthèses **extérieures** sont celles de `append( ... )`, comme dans `liste.append(22.4)` ;
- les parenthèses **intérieures** fabriquent le tuple `(moyenne, ville)`.

On ajoute donc **un seul** élément à la liste : un tuple qui contient deux valeurs.

```python
essai = []
essai.append(24.55, "Toulouse")
```

1. Dernière ligne : `TypeError: list.append() takes exactly one argument (2 given)`. Traduction : « `append`
   prend exactement une valeur (on lui en a donné 2) ».
2. Sans les parenthèses intérieures, la virgule sépare deux valeurs données à `append`.

✅ Correction : `essai.append((24.55, "Toulouse"))`.

On trie ensuite la liste de tuples, puis on récupère les noms avec une compréhension qui **déballe** chaque tuple :

```python
paires_triees = sorted(paires)
print(paires_triees)
noms = [ville for moyenne, ville in paires_triees]
print(noms)
print(f"La plus chaude : {noms[-1]}")
```

```python
records = {"Nîmes": 29.2, "Pau": 21.0, "Albi": 26.8}
paires = []
for ville, t in records.items():
    paires.append((t, ville))
print([ville for t, ville in sorted(paires)])
```

température est le premier élément de chaque tuple.

</details>

## 7. Compter et additionner par catégorie

C'est **le** geste du dictionnaire en data : regrouper par catégorie. Dans Excel, c'est un **tableau croisé
dynamique** (TCD). Exemple : combien de jours de chaque type de ciel ?

```python
ciel = ["soleil", "pluie", "soleil", "nuages", "soleil", "pluie"]
comptes = {}                                   # dictionnaire vide
for temps in ciel:
    comptes[temps] = comptes.get(temps, 0) + 1
print(comptes)
```

Décortiquons `comptes[temps] = comptes.get(temps, 0) + 1` :

1. `comptes.get(temps, 0)` : combien en a-t-on déjà compté ? 0 si c'est la première fois (la clé n'existe pas
   encore).
2. `+ 1` : on en ajoute un.
3. `comptes[temps] = ...` : on range le nouveau compte (la clé est créée la première fois, modifiée ensuite).

| Tour | `temps` | `comptes.get(temps, 0)` | `comptes` après |
|---|---|---|---|
| 1 | soleil | 0 | `{'soleil': 1}` |
| 2 | pluie | 0 | `{'soleil': 1, 'pluie': 1}` |
| 3 | soleil | 1 | `{'soleil': 2, 'pluie': 1}` |
| 4 | nuages | 0 | `{'soleil': 2, 'pluie': 1, 'nuages': 1}` |
| … | | | |

`{}` est un **dictionnaire vide**, le point de départ (comme `0` pour une somme, ou `[]` pour une liste).

Même geste pour **additionner** par catégorie : on ajoute la valeur au lieu de 1.

```python
villes = ["Toulouse", "Albi", "Toulouse", "Albi"]
pluies = [17.4, 9.4, 0.7, 5.0]
total_par_ville = {}
for i in range(len(villes)):
    ville = villes[i]
    total_par_ville[ville] = total_par_ville.get(ville, 0) + pluies[i]
print(total_par_ville)
```

## 8. La liste de dictionnaires : un tableau

Voici l'idée la plus importante du module. Un tableau Excel, c'est :

- des **lignes** : chaque ligne est un relevé ;
- des **colonnes**, avec un en-tête : `date`, `ville`, `temp_max`, `pluie_mm`.

En Python : **une ligne = un dictionnaire** (les clés sont les en-têtes), et **le tableau = une liste de ces
dictionnaires**.

| | date | ville | temp_max | pluie_mm |
|---|---|---|---|---|
| ligne 0 | 2026-09-01 | Toulouse | 25.9 | 17.4 |
| ligne 1 | 2026-09-01 | Albi | 26.8 | 9.4 |
| ligne 2 | 2026-09-02 | Toulouse | 23.2 | 0.7 |
| ligne 3 | 2026-09-02 | Albi | 27.0 | 5.0 |

```python
table = [
    {"date": "2026-09-01", "ville": "Toulouse", "temp_max": 25.9, "pluie_mm": 17.4},
    {"date": "2026-09-01", "ville": "Albi", "temp_max": 26.8, "pluie_mm": 9.4},
    {"date": "2026-09-02", "ville": "Toulouse", "temp_max": 23.2, "pluie_mm": 0.7},
    {"date": "2026-09-02", "ville": "Albi", "temp_max": 27.0, "pluie_mm": 5.0},
]
print(len(table))               # le nombre de lignes
print(table[1])                 # la ligne d'index 1 : un dictionnaire
print(table[1]["ville"])        # la cellule « ligne 1, colonne ville »
print(list(table[0].keys()))    # les en-têtes de colonnes
```

Lis `table[1]["ville"]` en deux temps, de gauche à droite :
1. `table[1]` : la liste donne la **ligne** d'index 1 (un dictionnaire) ;
2. `["ville"]` : le dictionnaire donne la valeur de la **colonne** `ville`.

C'est l'adresse d'une cellule, comme `C3` dans Excel : une ligne, une colonne.

🔮 **Prédis** : Avant d'exécuter : que va afficher cette cellule ? Écris ta réponse sur papier (ou dans ta tête), puis exécute.

```python
print(table[-1]["temp_max"])
print(table[0]["date"] == table[1]["date"])
```

<details><summary>👀 Voir l'explication</summary>

`27.0` (la dernière ligne, colonne `temp_max`), puis `True` (les deux premières lignes ont la même date).

</details>

### Parcourir les lignes

`for ligne in table:` donne un dictionnaire à chaque tour : « pour chaque ligne du tableau ».

```python
for ligne in table:
    print(f"{ligne['date']} | {ligne['ville']} | {ligne['temp_max']} °C")
```

### Extraire une colonne

Une colonne, c'est une liste : la même clé, lue dans chaque ligne. Une compréhension de liste (module 05) le fait
en une ligne. Ensuite, tous les outils du module 05 s'appliquent.

```python
maximales = [ligne["temp_max"] for ligne in table]
print(maximales)
print(f"Moyenne : {sum(maximales) / len(maximales):.2f} °C")
```

### Filtrer des lignes

Comme le filtre automatique d'Excel : on garde les lignes qui respectent une condition. Le résultat est encore une
table (une liste de dictionnaires).

```python
lignes_albi = [ligne for ligne in table if ligne["ville"] == "Albi"]
print(lignes_albi)
print(len(lignes_albi))
```

- Une **ligne** : `table.append({...})`, un nouveau dictionnaire au bout de la liste.
- Une **colonne calculée** : on parcourt les lignes et on ajoute une clé à chaque dictionnaire.

```python
table.append({"date": "2026-09-03", "ville": "Toulouse", "temp_max": 27.2, "pluie_mm": 0.0})
for ligne in table:
    ligne["temp_f"] = round(ligne["temp_max"] * 9 / 5 + 32, 1)
print(len(table))
print(table[-1])
```

### La moyenne par ville

On combine la section 7 (additionner par catégorie) et la table. Deux dictionnaires accumulateurs : la **somme**
par ville et le **nombre de lignes** par ville. Puis on divise, ville par ville.

```python
sommes = {}
nombres = {}
for ligne in table:
    ville = ligne["ville"]
    sommes[ville] = sommes.get(ville, 0) + ligne["temp_max"]
    nombres[ville] = nombres.get(ville, 0) + 1

moyennes = {}
for ville, somme in sommes.items():
    moyennes[ville] = round(somme / nombres[ville], 2)
print(moyennes)
```

C'est exactement ce que fait un tableau croisé dynamique « moyenne de `temp_max` par `ville` ». Au module 10,
**pandas** le fera en une ligne… et tu sauras ce qu'il fait à l'intérieur.

### Variante : regrouper les valeurs par ville

Au lieu d'additionner, on peut **ranger** les valeurs de chaque ville dans une liste. La valeur d'un dictionnaire
peut être n'importe quoi, même une liste :

```python
par_ville = {}
for ligne in table:
    ville = ligne["ville"]
    if ville not in par_ville:
        par_ville[ville] = []                    # première fois : une liste vide pour cette ville
    par_ville[ville].append(ligne["temp_max"])   # on ajoute à la liste de CETTE ville
print(par_ville)
print(max(par_ville["Toulouse"]))
```

## 9. Les ensembles (`set`) : dédoublonner

Quelles villes apparaissent dans la table ? La colonne `ville` contient des doublons. Un **ensemble** (`set` en
anglais) est une collection **sans doublons** : chaque valeur n'y apparaît qu'une fois. C'est le bouton
« Supprimer les doublons » d'Excel.

```python
colonne_villes = [ligne["ville"] for ligne in table]
print(colonne_villes)
villes_uniques = set(colonne_villes)
print(villes_uniques)
print(len(villes_uniques))
```

- `set(liste)` fabrique un ensemble à partir d'une liste.
- Un ensemble s'affiche entre accolades `{ }`, mais **sans** `clé:` : ce n'est pas un dictionnaire.
- Un ensemble **n'a pas d'ordre** : pas d'index (`villes_uniques[0]` plante), et l'ordre d'affichage peut
  changer. Pour une liste triée, on utilise `sorted()`, qui renvoie une liste.
- `in` marche, `len` aussi ; `.add(valeur)` ajoute un élément (sans effet s'il y est déjà).
- Attention : `{}` tout seul est un **dictionnaire** vide. Un ensemble vide s'écrit `set()`.

```python
villes_triees = sorted(villes_uniques)
print(villes_triees)

vues = set()
vues.add("Albi")
vues.add("Albi")             # déjà là : rien ne change
vues.add("Tarbes")
print(len(vues), "Albi" in vues)
```

### Repérer des lignes en double

Un ensemble peut contenir des **tuples**. Pour savoir si une ligne est un doublon, on regarde le couple
`(date, ville)` : s'il a déjà été vu, c'est un doublon.

Pourquoi un tuple et pas une liste ? Un ensemble range ses éléments d'après leur valeur, une fois pour toutes :
il n'accepte que des valeurs **qu'on ne peut pas modifier** (nombres, textes, tuples). Une liste, modifiable, est
refusée :

```python
deja_vus = set()
deja_vus.add(["2026-09-01", "Albi"])
```

1. Dernière ligne : `TypeError: unhashable type: 'list'`. *Unhashable* veut dire, en gros, « impossible à ranger
   dans un ensemble » (ni comme clé de dictionnaire), parce qu'une liste peut changer.
2. La flèche montre la liste `["2026-09-01", "Albi"]`.

✅ Correction : utilise un tuple, figé : `deja_vus.add(("2026-09-01", "Albi"))` (doubles parenthèses, comme pour
`append`).

```python
releves = [
    {"date": "2026-09-01", "ville": "Albi"},
    {"date": "2026-09-01", "ville": "Tarbes"},
    {"date": "2026-09-01", "ville": "Albi"},       # doublon !
]
deja_vus = set()
nb_doublons = 0
for ligne in releves:
    cle = (ligne["date"], ligne["ville"])
    if cle in deja_vus:
        nb_doublons = nb_doublons + 1
    else:
        deja_vus.add(cle)
print(f"{nb_doublons} doublon(s)")
```

## 🧠 Mini-quiz

**1.** `r = {"ville": "Albi", "pluie_mm": 9.4}`. Que donne `r["Albi"]` ?
<details><summary>Réponse</summary>

`KeyError: 'Albi'` : `"Albi"` est une **valeur**, pas une clé. La clé est `"ville"`.

<details><summary>Réponse</summary>

`0` : la clé n'existe pas, `.get` renvoie la valeur par défaut.
</details>

<details><summary>Réponse</summary>

La première ligne **modifie** la pluie (la clé existe), la seconde **ajoute** une rubrique (la clé n'existait pas).
</details>

la température de Tarbes ?
<details><summary>Réponse</summary>

`table[1]["temp_max"]` : la ligne d'index 1, puis la colonne `temp_max`.
</details>

**5.** Comment obtenir la liste triée, sans doublons, de `["Albi", "Tarbes", "Albi"]` ?
<details><summary>Réponse</summary>

`sorted(set(["Albi", "Tarbes", "Albi"]))` donne `['Albi', 'Tarbes']`.
</details>
