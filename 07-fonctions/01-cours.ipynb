## 1. Pourquoi des fonctions ?

Au module 06, tu as calculé la moyenne des températures de plusieurs villes. Le code ressemblait à ça :

````python
temps_toulouse = [25.9, 23.2, 27.2]
moyenne_toulouse = sum(temps_toulouse) / len(temps_toulouse)

temps_albi = [26.8, 27.0, 23.4]
moyenne_albi = sum(temps_albi) / len(temps_albi)

temps_tarbes = [22.0, 23.1, 21.4]
moyenne_tarbes = sum(temps_tarbes) / len(temps_tarbes)

print(round(moyenne_toulouse, 1), round(moyenne_albi, 1), round(moyenne_tarbes, 1))
````

Trois fois la **même formule**, recopiée. Si demain Inès veut la moyenne arrondie autrement, tu dois corriger
trois endroits (et tu en oublieras un).

Une **fonction**, c'est une **recette de cuisine** :
- elle a un **nom** (« pâte à crêpes ») ;
- elle attend des **ingrédients** (farine, œufs, lait) : ce sont ses **paramètres** ;
- elle suit des **étapes**, toujours les mêmes ;
- elle **rend un résultat** (la pâte) : c'est la valeur **renvoyée**.

Tu écris la recette **une fois**, puis tu la suis autant de fois que tu veux, avec des ingrédients différents.

Bonne nouvelle : tu utilises déjà des fonctions depuis le module 01 ! `print`, `len`, `round`, `sum`, `max`,
`type` sont des **fonctions intégrées** (fournies avec Python). Quand tu écris `len("Albi")`, tu **appelles**
la fonction `len` en lui donnant l'**argument** `"Albi"`, et elle te **renvoie** `4`.

## 2. Définir et appeler ta première fonction

Pour **définir** (écrire) une fonction, on utilise le mot `def` (pour *define*, définir) :

```python
def nom_de_la_fonction():
    instruction 1
    instruction 2
```

Lis bien chaque morceau :
- `def` puis le **nom** (mêmes règles que les variables : minuscules, `_` entre les mots, pas d'accent conseillé) ;
- des **parenthèses** `()` (vides pour l'instant) ;
- les **deux-points** `:` à la fin de la ligne, comme après un `if` ou un `for` ;
- le **bloc** : les lignes décalées de **4 espaces** (l'indentation, vue au module 03). C'est le corps de la recette.

````python
def afficher_entete():
    print("=== MétéoLocale ===")
    print("Bulletin de septembre")
````

Tu as exécuté la cellule... et **rien ne s'est affiché**. C'est normal ! Définir une fonction, c'est **écrire la
recette dans le cahier**, pas cuisiner. Python a seulement retenu : « `afficher_entete` = ces deux lignes ».

Pour **exécuter** la recette, on **appelle** la fonction : son nom **suivi de parenthèses**.

````python
afficher_entete()
````

## 3. Les paramètres : les ingrédients de la recette

Une recette qui fait toujours la même chose est vite limitée. On veut pouvoir lui donner des **ingrédients**.

Entre les parenthèses du `def`, on écrit des **paramètres** : des noms de variables qui seront remplis **au
moment de l'appel**. Les valeurs données à l'appel s'appellent des **arguments**.

Analogie : sur un formulaire, la case « Ville : ______ » est le **paramètre** ; ce que tu écris dedans
(« Albi ») est l'**argument**.

````python
def afficher_temperature(ville, temperature):
    print(f"{ville} : {temperature} °C")

afficher_temperature("Toulouse", 25.9)
afficher_temperature("Albi", 26.8)
````

Au premier appel, Python fait comme si tu avais écrit `ville = "Toulouse"` et `temperature = 25.9`, puis il
exécute le bloc. Au second appel, `ville = "Albi"` et `temperature = 26.8`.

Les arguments sont rangés **dans l'ordre** : le 1ᵉʳ argument va dans le 1ᵉʳ paramètre, le 2ᵉ dans le 2ᵉ.

Python ne comprend pas le sens des mots : il range **dans l'ordre**. `ville` reçoit `22.0` et `temperature`
reçoit `"Tarbes"`. Résultat : `22.0 : Tarbes °C`. Pas d'erreur, mais un affichage absurde. **L'ordre des
arguments compte.**

</details>

## 4. `return` : la fonction **renvoie** un résultat

Jusqu'ici nos fonctions **affichent**. Mais souvent on veut **récupérer** le résultat pour continuer à calculer :
le ranger dans une variable, l'additionner, le mettre dans une liste.

Le mot `return` (renvoyer) dit : « voici le résultat de la recette, je le rends à celui qui m'a appelé ».

````python
def ecart(temp_min, temp_max):
    return temp_max - temp_min

resultat = ecart(12.0, 25.9)
print(resultat)
````

Que s'est-il passé ?
1. `ecart(12.0, 25.9)` exécute le bloc avec `temp_min = 12.0` et `temp_max = 25.9` ;
2. `return temp_max - temp_min` calcule `13.9` et le **renvoie** ;
3. l'appel `ecart(12.0, 25.9)` est **remplacé par** `13.9`, et ce `13.9` est rangé dans `resultat`.

C'est exactement ce que fait `len("Albi")`, remplacé par `4`.

Python affiche `13.899999999999999` et non `13.9` : c'est la petite imprécision des nombres à virgule vue au
module 01. `round(resultat, 1)` l'arrondit à `13.9`.

🔮 **Prédis** : Avant d'exécuter : que va afficher cette cellule ? Écris ta réponse sur papier (ou dans ta tête), puis exécute.

````python
def ecart(temp_min, temp_max):
    return temp_max - temp_min

total = ecart(10.0, 20.0) + ecart(15.0, 18.0)
print(total)
````

<details><summary>👀 Voir l'explication</summary>

Chaque appel est remplacé par sa valeur : `ecart(10.0, 20.0)` vaut `10.0`, `ecart(15.0, 18.0)` vaut `3.0`.
Donc `total = 10.0 + 3.0`, et on affiche **13.0**. C'est tout l'intérêt de `return` : le résultat se réutilise
dans un calcul.

</details>

### `return` ou `print` ? La différence qui piège tout le monde

- `print` **montre** une valeur **à l'humain**, sur l'écran. Le programme, lui, ne la récupère pas.
- `return` **rend** une valeur **au programme**, qui peut la ranger et calculer avec. Rien ne s'affiche.

Analogie : `print`, c'est **lire le résultat à voix haute** ; `return`, c'est **poser le résultat sur le bureau
de ton collègue**. Pour qu'il puisse s'en servir, il faut le poser, pas seulement le dire.

Une fonction **sans** `return` renvoie quand même quelque chose : une valeur spéciale nommée **`None`**
(« rien » en anglais). `None` est la valeur qui veut dire « aucune valeur ».

🔮 **Prédis** : Avant d'exécuter : que va afficher cette cellule ? Écris ta réponse sur papier (ou dans ta tête), puis exécute.

````python
def ecart_affiche(temp_min, temp_max):
    print(temp_max - temp_min)

valeur = ecart_affiche(10.0, 20.0)
print("valeur contient :", valeur)
````

<details><summary>👀 Voir l'explication</summary>

Deux lignes s'affichent :
```
10.0
valeur contient : None
```
La première vient du `print` **à l'intérieur** de la fonction. Mais la fonction n'a pas de `return` : elle renvoie
`None`, et c'est `None` qui est rangé dans `valeur`. Le 10.0 a été **dit à voix haute**, pas **posé sur le bureau**.

</details>

### ⚠️ Erreur fréquente : calculer avec le résultat d'une fonction qui affiche au lieu de renvoyer

````python
def ecart_affiche(temp_min, temp_max):
    print(temp_max - temp_min)

total = ecart_affiche(10.0, 20.0) + 1
````

**Lecture** :
- Le `10.0` s'affiche d'abord (le `print` de la fonction a bien eu lieu).
- Dernière ligne : `TypeError: unsupported operand type(s) for +: 'NoneType' and 'int'`

- Traduction : « tu essaies de faire `None + 1` ». Quand tu vois `NoneType` dans une erreur, demande-toi :
  **quelle fonction a oublié son `return` ?**

**Correction** : remplacer `print(...)` par `return ...` dans la fonction.

````python
def ecart(temp_min, temp_max):
    return temp_max - temp_min

total = ecart(10.0, 20.0) + 1
print(total)
````

### `return` arrête la fonction

Dès que Python rencontre `return`, il **sort** de la fonction : les lignes suivantes du bloc ne sont pas
exécutées. On s'en sert avec un `if` pour renvoyer des résultats différents selon les cas.

````python
def classer_journee(temp_max):
    if temp_max >= 30:
        return "canicule"
    if temp_max >= 25:
        return "chaud"
    return "normal"

print(classer_journee(31.0))
print(classer_journee(26.8))
print(classer_journee(22.0))
````

### Renvoyer deux valeurs d'un coup

Au module 06, tu as vu les **tuples** (valeurs séparées par des virgules) et le **déballage**
(`a, b = (1, 2)`). Une fonction peut renvoyer un tuple, qu'on déballe à l'arrivée.

````python
def min_et_max(valeurs):
    return min(valeurs), max(valeurs)

plus_froid, plus_chaud = min_et_max([25.9, 23.2, 27.2, 24.1])
print(plus_froid, plus_chaud)
````

## 5. Des fonctions qui travaillent sur des listes et des dictionnaires

Un paramètre peut recevoir **n'importe quelle valeur** : un nombre, une chaîne, une liste, un dictionnaire, une
liste de dictionnaires. Voici enfin la recette de la moyenne, écrite **une seule fois** :

````python
def moyenne(valeurs):
    return sum(valeurs) / len(valeurs)

print(round(moyenne([25.9, 23.2, 27.2]), 1))
print(round(moyenne([26.8, 27.0, 23.4]), 1))
print(round(moyenne([22.0, 23.1, 21.4]), 1))
````

Compare avec la section 1 : même résultat, mais la formule n'est écrite **qu'à un seul endroit**. Si elle doit
changer, tu la changes une fois.

Le corps d'une fonction peut contenir tout ce que tu connais : boucles, conditions, listes...

````python
def compter_jours_pluie(pluies):
    compteur = 0
    for pluie in pluies:
        if pluie > 0:
            compteur = compteur + 1
    return compteur

print(compter_jours_pluie([17.4, 0.0, 0.0, 10.4, 0.0]))
````

🔮 **Prédis** : Avant d'exécuter : que va afficher cette cellule ? Écris ta réponse sur papier (ou dans ta tête), puis exécute.

````python
releves = [
    {"ville": "Toulouse", "temp_max": 25.9},
    {"ville": "Albi", "temp_max": 26.8},
    {"ville": "Toulouse", "temp_max": 23.2},
]

def temperatures_de(releves, ville):
    resultat = []
    for releve in releves:
        if releve["ville"] == ville:
            resultat.append(releve["temp_max"])
    return resultat

print(temperatures_de(releves, "Toulouse"))
print(temperatures_de(releves, "Perpignan"))
````

<details><summary>👀 Voir l'explication</summary>

```
[25.9, 23.2]
[]
```
La fonction parcourt la liste de dictionnaires (« pour chaque ligne du tableau ») et garde les `temp_max` de la
ville demandée. Pour Perpignan, aucune ligne ne correspond : la liste reste **vide**, `[]`.

</details>

### Le motif « ranger par ville » : un dictionnaire de listes

Au module 06, tu as **compté** par ville avec un dictionnaire `ville -> nombre`. Souvent, on veut plutôt **ranger**
les valeurs par ville : un dictionnaire `ville -> liste de valeurs`. C'est comme une armoire avec un tiroir par
ville : la première fois qu'on rencontre une ville, on **crée son tiroir vide** (`[]`) ; ensuite, on **ajoute**
dans son tiroir avec `append`.

🔮 **Prédis** : Avant d'exécuter : que va afficher cette cellule ? Écris ta réponse sur papier (ou dans ta tête), puis exécute.

````python
releves = [
    {"ville": "Toulouse", "temp_max": 25.9},
    {"ville": "Albi", "temp_max": 26.8},
    {"ville": "Toulouse", "temp_max": 23.2},
]

par_ville = {}
for releve in releves:
    ville = releve["ville"]
    if ville not in par_ville:
        par_ville[ville] = []                  # 1re fois : un tiroir vide
    par_ville[ville].append(releve["temp_max"])   # on range la valeur dans le tiroir de la ville

print(par_ville)
````

<details><summary>👀 Voir l'explication</summary>

`{'Toulouse': [25.9, 23.2], 'Albi': [26.8]}`. Au 1er tour, `Toulouse` n'est pas une clé : on crée `[]`, puis on
ajoute 25.9. Au 2e tour, même chose pour Albi. Au 3e tour, `Toulouse` existe déjà : on ajoute 23.2 **dans sa liste**.
Ensuite, il suffit d'appliquer `moyenne` à chaque liste (`for ville, valeurs in par_ville.items():`).
Sans le `if`, `par_ville[ville].append(...)` planterait au 1er tour avec un `KeyError` (module 06) : le tiroir n'existe pas encore.

</details>

## 6. Les valeurs par défaut

Parfois un paramètre a presque toujours la même valeur. On peut lui donner une **valeur par défaut** avec `=`
dans le `def` : si l'appel ne le précise pas, c'est elle qui sert.

Analogie : sur un bon de commande, « Quantité : 1 » est pré-rempli. Tu ne changes la case que si tu veux autre chose.

````python
def celsius_vers_fahrenheit(celsius, chiffres=1):
    return round(celsius * 9 / 5 + 32, chiffres)

print(celsius_vers_fahrenheit(25.9))      # chiffres vaut 1 (défaut)
print(celsius_vers_fahrenheit(25.9, 0))   # chiffres vaut 0
````

🔮 **Prédis** : Avant d'exécuter : que va afficher cette cellule ? Écris ta réponse sur papier (ou dans ta tête), puis exécute.

````python
def jours_au_dessus(temperatures, seuil=25):
    compteur = 0
    for t in temperatures:
        if t > seuil:
            compteur = compteur + 1
    return compteur

temps = [25.9, 23.2, 27.2, 26.1, 25.3]
print(jours_au_dessus(temps))
print(jours_au_dessus(temps, 26))
````

<details><summary>👀 Voir l'explication</summary>

```
4
2
```
Premier appel : `seuil` vaut `25` (défaut) → 25.9, 27.2, 26.1, 25.3 sont au-dessus : **4**.
Second appel : on donne `26`, qui remplace le défaut → 27.2 et 26.1 : **2**.

</details>

````python
def arrondir(chiffres=1, valeur):
    return round(valeur, chiffres)
````

## 7. Les arguments nommés

À l'appel, on peut écrire le **nom du paramètre** devant l'argument : `nom=valeur`. C'est un **argument nommé**.

Avantages :
- l'appel est plus **lisible** (on voit à quoi sert chaque valeur) ;
- l'**ordre** n'a plus d'importance pour les arguments nommés.

Tu en as peut-être déjà croisé un : `print("a", "b", sep=" - ")`, où `sep` (séparateur) est un argument nommé de `print`.

````python
def fiche(ville, temp_max, unite="°C"):
    return f"{ville} : {temp_max} {unite}"

print(fiche("Albi", 26.8))
print(fiche(temp_max=26.8, ville="Albi"))
print(fiche("Albi", 80.2, unite="°F"))
````

````python
def fiche(ville, temp_max, unite="°C"):
    return f"{ville} : {temp_max} {unite}"

print(fiche(unite="degrés", ville="Tarbes", temp_max=22.0))
````

<details><summary>👀 Voir l'explication</summary>

`Tarbes : 22.0 degrés`. Les trois arguments sont nommés : Python range chacun **dans le paramètre de même nom**,
quel que soit l'ordre d'écriture.

</details>

### ⚠️ Erreur fréquente : un argument normal après un argument nommé

````python
print(fiche(ville="Albi", 26.8))
````

**Lecture** : `SyntaxError: positional argument follows keyword argument`.
- *positional argument* = argument « par position » (sans nom) ;
- *keyword argument* = argument nommé (*keyword* = mot-clé) ;
- « un argument par position suit un argument nommé » : Python ne sait plus dans quelle case ranger `26.8`.

**Règle** : d'abord les arguments sans nom (dans l'ordre), ensuite les arguments nommés.
**Correction** : `fiche("Albi", temp_max=26.8)` ou `fiche(ville="Albi", temp_max=26.8)`.

## 8. La portée : ce qui se passe dans la fonction reste dans la fonction

Une variable créée **à l'intérieur** d'une fonction (y compris ses paramètres) est **locale** : elle n'existe que
pendant l'exécution de la fonction. Quand la fonction se termine, elle disparaît. On appelle ça la **portée**
(l'endroit du programme où un nom est visible).

Analogie : le **brouillon** sur lequel le cuisinier fait ses calculs est jeté à la fin de la recette. Ce qui
reste, c'est le plat qu'il te rend (le `return`).

````python
def total_pluie(pluies):
    somme = 0
    for p in pluies:
        somme = somme + p
    return somme

resultat = total_pluie([17.4, 0.7, 0.0])
print(round(resultat, 1))
````

### ⚠️ Erreur fréquente : utiliser une variable locale en dehors de sa fonction

````python
print(somme)
````

**Lecture** : `NameError: name 'somme' is not defined`. `somme` vivait **dans** `total_pluie` ; elle a disparu à
la fin de l'appel. Ce qui sort d'une fonction, c'est **uniquement** ce qu'elle renvoie avec `return`.

**Correction** : utiliser la variable qui a reçu le résultat, ici `resultat`.

🔮 **Prédis** : Avant d'exécuter : que va afficher cette cellule ? Écris ta réponse sur papier (ou dans ta tête), puis exécute.

````python
ville = "Toulouse"

def changer_ville():
    ville = "Albi"
    return ville

nouvelle = changer_ville()
print(ville, nouvelle)
````

<details><summary>👀 Voir l'explication</summary>

`Toulouse Albi`. Le `ville` **dans** la fonction est une **autre boîte**, locale, qui porte la même étiquette.
Modifier la boîte du brouillon ne touche pas à la boîte du dehors. Pour récupérer `"Albi"` dehors, on passe par
`return` (rangé dans `nouvelle`).

</details>

**Bonne pratique** : une fonction reçoit ce dont elle a besoin **par ses paramètres** et rend son travail **par
`return`**. Elle ne compte pas sur des variables du dehors. Ainsi, tu peux la comprendre (et la tester) toute seule.

## 9. La docstring : le mode d'emploi de la fonction

Juste sous la ligne `def`, on peut écrire une chaîne entre **triples guillemets** `"""..."""` (une chaîne qui
peut tenir sur plusieurs lignes) : c'est la **docstring** (*documentation string*). Elle explique **ce que fait**
la fonction, ce qu'elle **attend** et ce qu'elle **renvoie**. Python ne l'exécute pas : c'est pour les humains.

La fonction intégrée `help(...)` affiche ce mode d'emploi.

````python
def moyenne(valeurs):
    """Renvoie la moyenne d'une liste de nombres.

    valeurs : une liste non vide de nombres, par exemple [25.9, 23.2].
    """
    return sum(valeurs) / len(valeurs)

help(moyenne)
````

`help` marche aussi sur les fonctions intégrées : essaie `help(round)` dans une nouvelle cellule (le texte est en
anglais, mais tu y retrouveras `ndigits`, le nombre de chiffres après la virgule).

Dans VS Code, quand tu survoles le nom d'une fonction, c'est aussi sa docstring qui s'affiche.

## 10. Découper un problème en fonctions

Inès te demande : « Pour chaque ville, donne-moi la température max moyenne, en °C et en °F. »

Méthode de pro, en trois temps :
1. **Écris en français** les étapes : « garder les relevés d'une ville », « prendre les temp_max », « faire la
   moyenne », « convertir en °F », « fabriquer la phrase ».
2. **Une petite fonction par étape**, chacune testée seule sur un petit exemple.
3. **Une fonction chef d'orchestre** qui appelle les autres.

Chaque petite fonction est comme un **poste dans une cuisine** : un poste découpe, un poste cuit, le chef assemble.

````python
releves = [
    {"ville": "Toulouse", "temp_max": 25.9},
    {"ville": "Albi", "temp_max": 26.8},
    {"ville": "Toulouse", "temp_max": 23.2},
    {"ville": "Albi", "temp_max": 27.0},
]

def moyenne(valeurs):
    """Moyenne d'une liste non vide de nombres."""
    return sum(valeurs) / len(valeurs)

def celsius_vers_fahrenheit(celsius):
    """Convertit des degrés Celsius en degrés Fahrenheit."""
    return celsius * 9 / 5 + 32

def temperatures_de(releves, ville):
    """Liste des temp_max de la ville demandée."""
    resultat = []
    for releve in releves:
        if releve["ville"] == ville:
            resultat.append(releve["temp_max"])
    return resultat

def resume_ville(releves, ville):
    """Phrase de bulletin : moyenne des temp_max d'une ville en °C et en °F."""
    moy = moyenne(temperatures_de(releves, ville))
    return f"{ville} : {moy:.1f} °C ({celsius_vers_fahrenheit(moy):.1f} °F)"

print(resume_ville(releves, "Toulouse"))
print(resume_ville(releves, "Albi"))
````

🔮 **Prédis** : Avant d'exécuter : que va afficher cette cellule ? Écris ta réponse sur papier (ou dans ta tête), puis exécute.

````python
print(celsius_vers_fahrenheit(moyenne([10, 20, 30])))
````

<details><summary>👀 Voir l'explication</summary>

Python évalue **de l'intérieur vers l'extérieur** : d'abord `moyenne([10, 20, 30])` → `20.0` ; puis
`celsius_vers_fahrenheit(20.0)` → `20.0 * 9 / 5 + 32` = **68.0**. Le résultat d'une fonction devient l'argument
de la suivante, comme une chaîne de montage.

</details>

Chaque fonction fait **une seule chose**, porte un **nom qui dit ce qu'elle fait** (un verbe ou un résultat :
`moyenne`, `celsius_vers_fahrenheit`, `resume_ville`) et a sa docstring. C'est exactement ce que tu vas faire
dans le fil rouge.

## 📌 À retenir

| Geste | Exemple | À savoir |
|---|---|---|
| Définir | `def ecart(temp_min, temp_max):` | `def`, nom, parenthèses, `:` puis bloc indenté de 4 espaces |
| Appeler | `ecart(12.0, 25.9)` | toujours avec des parenthèses ; le `def` doit avoir été exécuté avant |
| Renvoyer | `return temp_max - temp_min` | l'appel est remplacé par la valeur renvoyée ; `return` arrête la fonction |
| `print` ≠ `return` | `print` montre à l'humain, `return` rend au programme | sans `return`, la fonction renvoie `None` |
| Deux résultats | `return mini, maxi` puis `a, b = f(...)` | un tuple, qu'on déballe |
| Valeur par défaut | `def f(valeur, chiffres=1):` | paramètres obligatoires d'abord |
| Argument nommé | `f(valeur=3.14159, chiffres=2)` | ordre libre ; les arguments sans nom d'abord |
| Portée locale | une variable créée dans `def` | disparaît à la fin de l'appel |
| Docstring | `"""Ce que fait la fonction."""` | sous le `def` ; `help(f)` l'affiche |

| Message d'erreur | Ce qu'il veut dire |
|---|---|
| `IndentationError: expected an indented block after function definition` | le bloc sous `def` n'est pas décalé |
| `NameError: name '...' is not defined` | faute de frappe, `def` pas exécuté, ou variable locale utilisée dehors |
| `TypeError: ...() missing 1 required positional argument: 'x'` | il manque l'argument `x` à l'appel |
| `TypeError: unsupported operand type(s) ... 'NoneType'` | une fonction a oublié son `return` |
| `SyntaxError: parameter without a default follows parameter with a default` | mets les paramètres obligatoires d'abord |
