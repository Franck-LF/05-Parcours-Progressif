## 2. Pourquoi fabriquer ses propres objets ?

Au module 06, un relevé était un dictionnaire. Ça marche, mais :
- rien n'empêche une faute de frappe dans une clé (`"temp_mx"`) ;
- les fonctions qui vont avec (`ecart(releve)`, `en_fahrenheit(releve)`) sont rangées **ailleurs**, loin des données.

```python
releve = {"date": "2026-09-01", "ville": "Toulouse", "temp_min": 12.0, "temp_max": 25.9}

def ecart(releve):
    return round(releve["temp_max"] - releve["temp_min"], 1)

print(ecart(releve))
```

Une **classe** regroupe au même endroit **les données** (les rubriques de la fiche) **et les fonctions** qui
travaillent dessus. C'est comme un formulaire qui aurait, imprimées au dos, les instructions pour calculer l'écart.

## 3. Écrire une classe : `class`, `__init__`, `self`

Voici notre premier formulaire, la classe `Releve`. Lis-le, on le décortique juste après.

```python
class Releve:
    def __init__(self, date, ville, temp_min, temp_max):
        self.date = date
        self.ville = ville
        self.temp_min = temp_min
        self.temp_max = temp_max


r1 = Releve("2026-09-01", "Toulouse", 12.0, 25.9)
print(r1.ville, r1.temp_max)
```

Ligne par ligne :

- `class Releve:` : on déclare une **classe** nommée `Releve`. Par convention, un nom de classe commence par une
  **majuscule** (`Releve`, `DataFrame`), alors qu'une variable s'écrit en minuscules (`r1`). Les deux-points et
  l'indentation, tu les connais (modules 03 et 07).
- `def __init__(self, ...):` : une fonction **à l'intérieur** de la classe. Son nom, `__init__` (deux tirets bas de
  chaque côté, pour *initialiser*), est spécial : Python l'appelle **tout seul** quand on fabrique un objet. C'est
  le moment où l'on **remplit la fiche**.
- `self` : le **premier paramètre**, toujours. Il désigne **la fiche en train d'être remplie**, l'objet lui-même
  (*self* = « soi-même » en anglais). Tu ne le donnes jamais toi-même : Python s'en charge.
- `self.ville = ville` : on range la valeur reçue dans une **rubrique** de la fiche. À gauche, `self.ville` est un
  **attribut** de l'objet ; à droite, `ville` est le paramètre reçu (une variable locale, module 07).
- `r1 = Releve("2026-09-01", "Toulouse", 12.0, 25.9)` : on **fabrique un objet**, en appelant la classe comme une
  fonction. Python crée une fiche vide, appelle `__init__` avec `self` = cette fiche, puis la range dans `r1`.
- `r1.ville` : on lit un attribut avec un **point**. Comme `meteo.shape` au module 10.

Un seul formulaire, autant de fiches qu'on veut. Chaque objet garde **ses propres** valeurs.

```python
class Releve:
    def __init__(self, date, ville, temp_min, temp_max):
        self.date = date
        self.ville = ville
        self.temp_min = temp_min
        self.temp_max = temp_max


r1 = Releve("2026-09-01", "Toulouse", 12.0, 25.9)
r2 = Releve("2026-09-01", "Tarbes", 10.4, 22.0)
print(r1.ville, r1.temp_max)
print(r2.ville, r2.temp_max)

r2.temp_max = 22.5          # on peut modifier une rubrique, comme une clé de dictionnaire
print(r2.temp_max, r1.temp_max)
```

## 4. Les méthodes : les fonctions de l'objet

Une **méthode** est une fonction écrite **dans** la classe. Son premier paramètre est toujours `self` : la méthode
peut ainsi lire les attributs de **son** objet.

```python
class Releve:
    def __init__(self, date, ville, temp_min, temp_max):
        self.date = date
        self.ville = ville
        self.temp_min = temp_min
        self.temp_max = temp_max

    def ecart(self):
        return round(self.temp_max - self.temp_min, 1)

    def en_fahrenheit(self):
        return round(self.temp_max * 9 / 5 + 32, 1)


r = Releve("2026-09-01", "Toulouse", 12.0, 25.9)
print(r.ecart())
print(r.en_fahrenheit())
```

- La méthode est **indentée** dans la classe, au même niveau que `__init__`. Une ligne vide entre deux méthodes,
  pour respirer.
- On l'appelle avec un **point** et des **parenthèses** : `r.ecart()`, comme `meteo.head()`.
- On ne donne rien entre les parenthèses, et pourtant `ecart` a un paramètre `self` : Python passe **l'objet placé
  avant le point**. `r.ecart()` revient à `Releve.ecart(r)`.

Comparé à la version dictionnaire de la section 2 : la fonction est **rangée avec ses données**.

## 5. Un bel affichage : `__repr__`

Que se passe-t-il si on affiche l'objet entier ?

```python
class Releve:
    def __init__(self, date, ville, temp_min, temp_max):
        self.date = date
        self.ville = ville
        self.temp_min = temp_min
        self.temp_max = temp_max


r = Releve("2026-09-01", "Toulouse", 12.0, 25.9)
print(r)
```

`<__main__.Releve object at 0x...>` : Python dit seulement « un objet Releve, rangé à telle adresse de la mémoire ».
Illisible.

La méthode spéciale `__repr__` (pour *représentation*, encore deux tirets bas de chaque côté) dit à Python
**comment afficher** l'objet. Elle doit **renvoyer une chaîne** (avec `return`, pas `print`). Python l'appelle tout
seul avec `print(r)`, et aussi quand l'objet est dans une liste.

```python
class Releve:
    def __init__(self, date, ville, temp_min, temp_max):
        self.date = date
        self.ville = ville
        self.temp_min = temp_min
        self.temp_max = temp_max

    def __repr__(self):
        return f"Releve({self.ville}, {self.date}, {self.temp_min}-{self.temp_max} °C)"


r = Releve("2026-09-01", "Toulouse", 12.0, 25.9)
print(r)
print([r, Releve("2026-09-01", "Albi", 12.6, 26.8)])
```

La fonction `repr(objet)` renvoie **le même texte** que celui que `print(objet)` affiche : c'est Python qui appelle
`__repr__` pour toi. Pratique pour ranger ce texte dans une variable.

```python
class Releve:
    def __init__(self, ville, temp_max):
        self.ville = ville
        self.temp_max = temp_max

    def __repr__(self):
        return f"Releve({self.ville}, {self.temp_max} °C)"


texte = repr(Releve("Albi", 26.8))
print(texte)
print(type(texte))
```

## 7. Un objet qui contient d'autres objets : la `Station`

Une station météo, c'est un nom de ville **et** la liste de ses relevés. Ses méthodes calculent des résumés. Un
attribut peut très bien être une liste, vide au départ, qu'une méthode remplit.

```python
class Releve:
    def __init__(self, date, ville, temp_min, temp_max, pluie_mm):
        self.date = date
        self.ville = ville
        self.temp_min = temp_min
        self.temp_max = temp_max
        self.pluie_mm = pluie_mm

    def ecart(self):
        return round(self.temp_max - self.temp_min, 1)

    def __repr__(self):
        return f"Releve({self.ville}, {self.date}, {self.temp_min}-{self.temp_max} °C)"


class Station:
    def __init__(self, nom):
        self.nom = nom
        self.releves = []            # la liste démarre vide

    def ajouter(self, releve):
        self.releves.append(releve)

    def nb_releves(self):
        return len(self.releves)

    def moyenne_max(self):
        total = 0
        for releve in self.releves:
            total = total + releve.temp_max
        return round(total / len(self.releves), 1)

    def __repr__(self):
        return f"Station({self.nom}, {self.nb_releves()} relevés)"


toulouse = Station("Toulouse")
toulouse.ajouter(Releve("2026-09-01", "Toulouse", 12.0, 25.9, 17.4))
toulouse.ajouter(Releve("2026-09-02", "Toulouse", 13.8, 23.2, 0.7))
print(toulouse)
print(toulouse.moyenne_max())
```

À noter :
- `self.releves = []` : chaque station a **sa propre** liste.
- Une méthode peut appeler une autre méthode du même objet avec `self.` : `self.nb_releves()` dans `__repr__`.
- `releve.temp_max` dans la boucle : on lit l'attribut d'un **autre** objet, le relevé.

```python
class Releve:
    def __init__(self, date, ville, temp_min, temp_max, pluie_mm):
        self.date = date
        self.ville = ville
        self.temp_min = temp_min
        self.temp_max = temp_max
        self.pluie_mm = pluie_mm

    def ecart(self):
        return round(self.temp_max - self.temp_min, 1)

    def __repr__(self):
        return f"Releve({self.ville}, {self.date}, {self.temp_min}-{self.temp_max} °C)"


class Station:
    def __init__(self, nom):
        self.nom = nom
        self.releves = []

    def ajouter(self, releve):
        self.releves.append(releve)

    def nb_releves(self):
        return len(self.releves)


albi = Station("Albi")
tarbes = Station("Tarbes")
albi.ajouter(Releve("2026-09-01", "Albi", 12.6, 26.8, 9.4))
albi.ajouter(Releve("2026-09-02", "Albi", 14.6, 27.0, 5.0))
print(albi.nb_releves(), tarbes.nb_releves())
```

## 8. Plus court : `@dataclass`

Écrire `self.date = date`, `self.ville = ville`… pour chaque rubrique, c'est répétitif. Le module standard
`dataclasses` (module 09 : `from … import …`) fournit `dataclass`, qui **écrit `__init__` et `__repr__` à ta place**.

On le pose **au-dessus** de la classe avec un `@` : on appelle ça un **décorateur**, une étiquette qui modifie ce
qui est écrit juste en dessous. Puis on liste les rubriques avec leur type : `ville: str` se lit « une rubrique
`ville`, qui contient du texte ». Python ne vérifie pas ces types : c'est une information pour le lecteur (et pour
`dataclass`, qui a besoin de la liste des rubriques).

```python
from dataclasses import dataclass


@dataclass
class Releve:
    date: str
    ville: str
    temp_min: float
    temp_max: float

    def ecart(self):
        return round(self.temp_max - self.temp_min, 1)


r = Releve("2026-09-01", "Toulouse", 12.0, 25.9)
print(r)
print(r.ville, r.ecart())
```

Même objet qu'à la section 4, sans écrire `__init__` ni `__repr__`. L'affichage automatique montre toutes les
rubriques : `Releve(date='2026-09-01', ville='Toulouse', temp_min=12.0, temp_max=25.9)`. On peut toujours ajouter
ses méthodes (`ecart`) en dessous des rubriques.

Bonus : deux dataclasses aux rubriques identiques sont **égales** avec `==`. Avec une classe écrite à la main, `==`
répond `False`, même pour deux fiches identiques.

## 9. Le lien avec l'IA : un modèle, c'est un objet avec `fit` et `predict`

Plus tard dans la formation, tu utiliseras **scikit-learn**, la bibliothèque d'apprentissage automatique la plus
répandue. Tu n'en as pas besoin aujourd'hui : on va construire **notre propre modèle**, tout petit, pour comprendre
son mode d'emploi. Il tient en deux méthodes :
- `fit` (« ajuster ») : le modèle **apprend** à partir de données passées. On dit aussi **entraîner** le modèle.
- `predict` (« prédire ») : le modèle **prévoit** des valeurs nouvelles avec ce qu'il a appris.

Un **modèle**, ici, est une recette de prévision. La plus simple qui soit : « demain, il fera la **moyenne** des jours
passés ». C'est naïf, mais c'est un vrai point de départ, que les data scientists utilisent pour comparer.

```python
class ModeleMoyenne:
    def __init__(self):
        self.moyenne = None

    def fit(self, valeurs):
        self.moyenne = round(sum(valeurs) / len(valeurs), 1)

    def predict(self, nb_jours):
        previsions = []
        for jour in range(nb_jours):
            previsions.append(self.moyenne)
        return previsions


modele = ModeleMoyenne()
print(modele.moyenne)          # rien appris pour l'instant

modele.fit([25.9, 23.2, 27.2, 26.1])
print(modele.moyenne)          # appris : la moyenne des valeurs

print(modele.predict(3))       # prévision pour les 3 prochains jours
```

- `__init__` met `self.moyenne = None` : `None` (module 07) veut dire « pas encore de valeur ». Le modèle n'a encore
  rien appris.
- `fit(valeurs)` **calcule** et **range** ce qu'il a appris dans un attribut. Il ne renvoie rien : il modifie le
  modèle.
- `predict(nb_jours)` **utilise** ce qui a été appris et **renvoie** une liste de prévisions. Ici, la même valeur
  pour chaque jour : notre modèle n'est pas très imaginatif.

Ce découpage `fit` puis `predict`, c'est **exactement** celui de scikit-learn : `modele.fit(...)`, puis
`modele.predict(...)`. Tu reconnaîtras le geste.

### Évaluer le modèle : apprendre sur le passé, tester sur la fin

Comment savoir si un modèle est bon ? On le fait **apprendre** sur les 23 premiers jours de septembre à Toulouse,
on lui fait **prévoir** les 7 derniers, et on compare à ce qui s'est vraiment passé. Les 7 derniers jours sont le
**jeu de test** : le modèle ne les a jamais vus.

Pour mesurer l'erreur d'un jour, on prend l'écart entre la prévision et la réalité, **sans son signe** : se tromper
de 2 degrés trop haut ou trop bas, c'est pareil. La fonction `abs()` (*absolute*, valeur absolue) enlève le signe :
`abs(-2.0)` vaut `2.0`, `abs(2.0)` vaut `2.0`. L'**erreur moyenne** est la moyenne de ces écarts.

```python
import pandas as pd

class ModeleMoyenne:
    def __init__(self):
        self.moyenne = None

    def fit(self, valeurs):
        self.moyenne = round(sum(valeurs) / len(valeurs), 1)

    def predict(self, nb_jours):
        previsions = []
        for jour in range(nb_jours):
            previsions.append(self.moyenne)
        return previsions


meteo = pd.read_csv("../donnees/meteo.csv")
temperatures = list(meteo[meteo["ville"] == "Toulouse"]["temp_max"])
entrainement = temperatures[:23]      # les 23 premiers jours : pour apprendre
test = temperatures[23:]              # les 7 derniers : pour vérifier

modele = ModeleMoyenne()
modele.fit(entrainement)
previsions = modele.predict(len(test))

total_erreurs = 0
for i in range(len(test)):
    total_erreurs = total_erreurs + abs(previsions[i] - test[i])
erreur_moyenne = round(total_erreurs / len(test), 2)

print("Appris :", modele.moyenne)
print("Prévu  :", previsions)
print("Réel   :", test)
print("Erreur moyenne :", erreur_moyenne, "°C")
```

### Un deuxième modèle, même mode d'emploi

Autre recette naïve : « demain, il fera comme **le dernier jour** connu ». On écrit une **autre classe**, avec les
**mêmes noms de méthodes**. Du coup, le code qui entraîne et évalue n'a pas à changer : on peut **échanger les
modèles** comme on change de cartouche. C'est la grande force de ce mode d'emploi commun.

```python
import pandas as pd

class ModeleMoyenne:
    def __init__(self):
        self.moyenne = None

    def fit(self, valeurs):
        self.moyenne = round(sum(valeurs) / len(valeurs), 1)

    def predict(self, nb_jours):
        previsions = []
        for jour in range(nb_jours):
            previsions.append(self.moyenne)
        return previsions


class ModeleDernierJour:
    def __init__(self):
        self.dernier = None

    def fit(self, valeurs):
        self.dernier = valeurs[-1]

    def predict(self, nb_jours):
        previsions = []
        for jour in range(nb_jours):
            previsions.append(self.dernier)
        return previsions


meteo = pd.read_csv("../donnees/meteo.csv")
temperatures = list(meteo[meteo["ville"] == "Toulouse"]["temp_max"])
entrainement = temperatures[:23]
test = temperatures[23:]

for modele in [ModeleMoyenne(), ModeleDernierJour()]:
    modele.fit(entrainement)
    previsions = modele.predict(len(test))
    total = 0
    for i in range(len(test)):
        total = total + abs(previsions[i] - test[i])
    print(modele.__class__.__name__, ":", round(total / len(test), 2), "°C d'erreur moyenne")
```

`modele.__class__.__name__` donne le nom de la classe de l'objet, sous forme de texte : pratique pour l'affichage,
tu n'auras pas à l'écrire toi-même.

Ici la moyenne se trompe de 1.8 °C en moyenne, le dernier jour de 1.9 °C : sur cette semaine, la
moyenne gagne de peu. Les vrais modèles de scikit-learn sont bien plus malins, mais ils s'utilisent **exactement**
comme ça.

⚠️ **Erreur fréquente : prédire avant d'avoir entraîné.**

```python
class ModeleMoyenne:
    def __init__(self):
        self.moyenne = None

    def fit(self, valeurs):
        self.moyenne = round(sum(valeurs) / len(valeurs), 1)

    def predict(self, nb_jours):
        previsions = []
        for jour in range(nb_jours):
            previsions.append(self.moyenne)
        return previsions


modele = ModeleMoyenne()
previsions = modele.predict(2)
print(previsions)
erreur = abs(previsions[0] - 24.7)
```

**Lisons le message.** D'abord, `print` a affiché `[None, None]` : le modèle a « prévu » du vide. Puis la dernière
ligne : `TypeError: unsupported operand type(s) for -: 'NoneType' and 'float'`.

- Traduction : « types non pris en charge pour la soustraction : `NoneType` et `float` ». `NoneType` est le type de
  `None` : on ne peut pas soustraire un nombre de « rien ».
- Pourquoi `None` ? Parce que `self.moyenne` vaut encore `None` : `fit` n'a jamais été appelé.
- La flèche montre la ligne `erreur = abs(...)`, mais la **vraie** cause est plus haut : l'oubli de `fit`. Quand une
  valeur est bizarre (`None`), remonte d'où elle vient.

**Correction** : toujours `modele.fit(...)` **avant** `modele.predict(...)`. scikit-learn a la même règle, avec un
message plus clair (« not fitted yet » : pas encore entraîné).

## 📌 À retenir

| Mot | Ce que c'est | Exemple |
|---|---|---|
| objet | une valeur qui range des informations et sait faire des choses | `"albi"`, `meteo`, `r` |
| classe | le formulaire vierge qui fabrique des objets | `class Releve:` |
| `__init__` | appelée toute seule à la fabrication : remplit la fiche | `def __init__(self, ville):` |
| `self` | l'objet lui-même ; premier paramètre de **toute** méthode | `self.ville = ville` |
| attribut | une information rangée dans l'objet, lue avec un point | `r.temp_max` |
| méthode | une fonction rangée dans la classe, appelée avec `()` | `r.ecart()` |
| `__repr__` | renvoie le texte affiché par `print(objet)` | `return f"Releve({self.ville})"` |
| `@dataclass` | écrit `__init__`, `__repr__` et `==` à ta place | `ville: str` |
| `fit` / `predict` | apprendre sur le passé / prévoir ; toujours `fit` d'abord | `modele.fit(donnees)` |

Erreurs : `TypeError … missing 1 required positional argument` = il manque une valeur à la fabrication ;
`AttributeError … has no attribute` = attribut mal écrit (ou classe sans parenthèses) ;
`TypeError … takes 0 positional arguments but 1 was given` = `self` oublié dans la méthode.
