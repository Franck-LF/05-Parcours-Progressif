## 1. Pourquoi répéter ?

Inès, ta responsable chez MétéoLocale, veut afficher les 30 jours de septembre. Avec ce que tu sais déjà, tu
devrais écrire 30 fois `print` :

````python
print("Jour 1")
print("Jour 2")
print("Jour 3")
# ... et encore 27 lignes presque identiques !
````

C'est long, ennuyeux, et une faute de frappe est vite arrivée. Dans Excel, tu ne retapes pas une formule sur
chaque ligne : tu la **tires vers le bas**, et Excel la répète pour chaque ligne du tableau.

En Python, l'outil qui répète s'appelle une **boucle** : un morceau de code que Python exécute plusieurs fois
de suite. Chaque passage dans la boucle s'appelle un **tour** (on dit aussi une *itération*).

Voici la même chose avec une boucle. Ne cherche pas encore à tout comprendre, regarde juste la longueur :

````python
for jour in range(1, 31):
    print(f"Jour {jour}")
````

Deux lignes au lieu de trente. Tout le module sert à comprendre ces deux lignes, morceau par morceau.

## 2. La boucle `for` sur une chaîne

On commence par le cas le plus simple : parcourir un texte **lettre par lettre**.

````python
ville = "Albi"
for lettre in ville:
    print(lettre)
````

Lis cette boucle comme une phrase : « **pour chaque** `lettre` **dans** `ville`, affiche la lettre ».

| Morceau | Rôle |
|---|---|
| `for` | le mot qui annonce la boucle (« pour » en anglais) |
| `lettre` | la **variable de boucle** : une étiquette que Python recolle sur un nouveau caractère à chaque tour. Tu choisis son nom. |
| `in` | « dans » |
| `ville` | ce qu'on parcourt : ici la chaîne `"Albi"` |
| `:` | les deux-points, obligatoires, comme après un `if` (module 03) |
| `    print(lettre)` | le **corps** de la boucle : les lignes **indentées** (décalées de 4 espaces) sous le `for`. Elles sont répétées à chaque tour. |

Que se passe-t-il, tour par tour ?

| Tour | Valeur de `lettre` | Ce qui s'affiche |
|---|---|---|
| 1 | `"A"` | `A` |
| 2 | `"l"` | `l` |
| 3 | `"b"` | `b` |
| 4 | `"i"` | `i` |

Quand il n'y a plus de lettre, la boucle s'arrête et Python passe à la ligne suivante, non indentée.

🔮 **Prédis** : Avant d'exécuter : que va afficher cette cellule ? Écris ta réponse sur papier (ou dans ta tête), puis exécute.

````python
for lettre in "Nîmes":
    print(lettre)
print("Fin")
````

<details><summary>👀 Voir l'explication</summary>

Cinq lignes `N`, `î`, `m`, `e`, `s`, puis `Fin` **une seule fois**.

`print("Fin")` n'est pas indenté : il ne fait pas partie du corps de la boucle. Il est exécuté une fois, après
le dernier tour. C'est l'indentation, et rien d'autre, qui dit à Python ce qui est répété.

</details>

Le corps de la boucle peut contenir plusieurs lignes, et même un `if` (module 03). Chaque ligne indentée est
exécutée à chaque tour, dans l'ordre :

````python
for lettre in "Tarbes":
    majuscule = lettre.upper()
    print(f"{lettre} devient {majuscule}")
````

## 3. `range()` : compter des tours

Pour parcourir les 30 jours de septembre, on n'a pas de texte à parcourir : on veut juste des **numéros**.
`range()` (« étendue » en anglais) fabrique une suite de nombres entiers que la boucle reçoit un par un.

### `range(fin)` : de 0 jusqu'à fin, fin **non comprise**

````python
for i in range(4):
    print(i)
````

`range(4)` donne `0`, `1`, `2`, `3` : **quatre** nombres, en commençant à **0**, et le `4` n'est pas compris.

En Python, on compte souvent à partir de 0. Pense à un ascenseur : le rez-de-chaussée est l'étage 0.
Le nom `i` est une habitude des programmeurs pour un simple numéro de tour.

Avec deux nombres, le premier est le **départ** (compris) et le second la **fin** (non comprise). Pour les

Moyen mnémotechnique : c'est comme un magasin ouvert « de 9 h à 12 h ». À 12 h pile, c'est fermé.

````python
for jour in range(1, 8):
    print(f"Jour {jour} de la première semaine")
````

### `range(debut, fin, pas)` : sauter de plusieurs en plusieurs

Un troisième nombre donne le **pas** : de combien on avance à chaque tour.

````python
for heure in range(0, 24, 6):
    print(f"Relevé à {heure} h")
````

On peut **calculer** avec la variable de boucle, comme avec n'importe quelle variable :

````python
for celsius in range(0, 31, 10):
    fahrenheit = celsius * 9 / 5 + 32
    print(f"{celsius} °C = {fahrenheit:.1f} °F")
````

### Calculer un index avec la variable de boucle

Au module 02, tu as pris un caractère avec `texte[0]` et un morceau avec `texte[0:2]`, avec des nombres écrits en
dur. Entre les crochets, on peut aussi mettre **une variable**, ou **un calcul** : Python calcule d'abord les
nombres, puis prend le morceau.

Exemple : le capteur colle des températures de **deux chiffres** : `"242731"` veut dire 24, 27, 31. Avec un pas
de 2, `i` vaut 0, 2, 4 : le début de chaque température. Le morceau de deux caractères qui commence à `i`, c'est
`temps[i:i + 2]`.

````python
temps = "242731"
for i in range(0, len(temps), 2):
    morceau = temps[i:i + 2]
    print(f"i vaut {i} : temps[{i}:{i + 2}] donne {morceau}")
````

| Tour | `i` | `i + 2` | `temps[i:i + 2]` devient | morceau |
|---|---|---|---|---|
| 1 | 0 | 2 | `temps[0:2]` | `"24"` |
| 2 | 2 | 4 | `temps[2:4]` | `"27"` |
| 3 | 4 | 6 | `temps[4:6]` | `"31"` |

`len(temps)` vaut 6 : `range(0, 6, 2)` donne bien 0, 2, 4. Le morceau est un texte : pour calculer, on le convertit
avec `int(morceau)`.

🔮 **Prédis** : Avant d'exécuter : que va afficher cette cellule ? Écris ta réponse sur papier (ou dans ta tête), puis exécute.

````python
ville = "Tarbes"
for i in range(0, 6, 3):
    print(ville[i:i + 3])
````

`Tar` puis `bes`. Premier tour : `i` vaut 0, donc `ville[0:3]`. Deuxième tour : `i` vaut 3, donc `ville[3:6]`.

</details>

````python
for jour in 30:
    print(jour)
````

## 4. L'accumulateur : faire une somme

Une boucle devient vraiment utile quand elle **calcule** au fil des tours. Exemple : la pluie tombée sur
5 jours. Le capteur l'envoie sous forme de texte, **un chiffre par jour**, en millimètres :
`"30120"` veut dire 3 mm, 0 mm, 1 mm, 2 mm, 0 mm.

Pour faire le total, on utilise un **accumulateur** : une variable qui grossit à chaque tour, comme une
tirelire où l'on glisse une pièce à chaque passage. C'est ce que fait la formule `SOMME` d'Excel, mais à la main.

````python
pluies = "30120"
total = 0                          # 1. AVANT la boucle : la tirelire est vide
for chiffre in pluies:
    total = total + int(chiffre)   # 2. DANS la boucle : on ajoute la valeur du tour
print(f"Total : {total} mm")       # 3. APRÈS la boucle : on lit le résultat
````

Trois temps, **toujours les mêmes** :

1. **Initialiser** (donner sa valeur de départ) avant la boucle : `total = 0`.
2. **Mettre à jour** dans la boucle : `total = total + ...`. Lis-le de droite à gauche : « prends l'ancien
   total, ajoute la valeur, range le résultat dans `total` ».
3. **Utiliser** le résultat après la boucle (ligne non indentée).

`chiffre` est un caractère (`"3"`, un `str`) : on le convertit avec `int(chiffre)` avant de l'additionner,
comme au module 02.

| Tour | `chiffre` | calcul | `total` après le tour |
|---|---|---|---|
| départ | | | 0 |
| 1 | `"3"` | 0 + 3 | 3 |
| 2 | `"0"` | 3 + 0 | 3 |
| 3 | `"1"` | 3 + 1 | 4 |
| 4 | `"2"` | 4 + 2 | 6 |
| 5 | `"0"` | 6 + 0 | 6 |

🔮 **Prédis** : Avant d'exécuter : que va afficher cette cellule ? Écris ta réponse sur papier (ou dans ta tête), puis exécute.

````python
total = 0
for jour in range(1, 5):
    total = total + jour
print(total)
````

<details><summary>👀 Voir l'explication</summary>

`10`. Les tours ajoutent 1, puis 2, puis 3, puis 4 : 0 + 1 + 2 + 3 + 4 = 10. Le `5` n'est pas compris dans
`range(1, 5)`.

</details>

### Le raccourci `+=`

`total = total + x` s'écrit si souvent que Python a un raccourci : `total += x`. Les deux lignes font
**exactement** la même chose. De même, `reserve -= 15` veut dire `reserve = reserve - 15`.
Utilise la forme qui te paraît la plus claire.

````python
vents = "4210"
total_vent = 0
for chiffre in vents:
    total_vent += int(chiffre)     # même chose que total_vent = total_vent + int(chiffre)
print(f"Total des vents : {total_vent}")
````

✅ Correction : écris `somme_pluie = 0` **avant** la ligne `for`.

## 5. Le compteur, la moyenne et le record

### Le compteur : combien de fois ?

Un **compteur** est un accumulateur qui augmente de 1 à chaque fois qu'une condition est vraie. C'est la formule
`NB.SI` d'Excel. Exemple : le ciel de la semaine, codé une lettre par jour : `S` soleil, `P` pluie,
`N` nuages. Combien de jours de pluie ?

````python
ciel = "SPPNSPS"
nb_pluie = 0
for jour in ciel:
    if jour == "P":
        nb_pluie = nb_pluie + 1
print(f"Jours de pluie : {nb_pluie}")
````

Remarque les **deux niveaux d'indentation** : le `if` est dans la boucle (4 espaces), et la ligne
`nb_pluie = nb_pluie + 1` est dans le `if` (8 espaces). Elle n'est exécutée que les tours où le jour vaut `"P"`.

🔮 **Prédis** : Avant d'exécuter : que va afficher cette cellule ? Écris ta réponse sur papier (ou dans ta tête), puis exécute.

````python
nb_pairs = 0
for jour in range(1, 11):
    if jour % 2 == 0:
        nb_pairs += 1
print(nb_pairs)
````

<details><summary>👀 Voir l'explication</summary>

`5`. `jour % 2` est le reste de la division par 2 (module 01) : il vaut 0 pour les nombres pairs. Entre 1 et 10,
les pairs sont 2, 4, 6, 8, 10 : cinq fois, le compteur augmente.

</details>

### La moyenne : un total et un compteur

La moyenne, c'est le total divisé par le nombre de valeurs. On fait **les deux accumulateurs dans la même
boucle**, puis on divise **après** la boucle.

````python
vents = "3425"           # force du vent, un chiffre par jour
total = 0
nb_jours = 0
for chiffre in vents:
    total = total + int(chiffre)
    nb_jours = nb_jours + 1
moyenne = total / nb_jours
print(f"{nb_jours} jours, total {total}, moyenne {moyenne:.1f}")
````

Ici on aurait pu utiliser `len(vents)` pour le nombre de jours. Mais le compteur devient indispensable quand on
ne garde **que certains** tours (on le verra avec `continue`).

### Le record : garder le plus grand

Pour trouver la valeur la plus haute, on garde dans une variable le **record vu jusqu'ici**. À chaque tour, si
la nouvelle valeur bat le record, elle le remplace. C'est comme un tableau des records sportifs : on ne
l'efface que lorsqu'un athlète fait mieux.

````python
vents = "3815924"
record = 0                   # plus petit que toutes les valeurs possibles
for chiffre in vents:
    valeur = int(chiffre)
    if valeur > record:
        record = valeur
print(f"Vent le plus fort : {record}")
````

🔮 **Prédis** : Avant d'exécuter : que va afficher cette cellule ? Écris ta réponse sur papier (ou dans ta tête), puis exécute.

````python
record = 0
for chiffre in "2741":
    valeur = int(chiffre)
    if valeur > record:
        record = valeur
    print(f"vu {valeur}, record {record}")
````

<details><summary>👀 Voir l'explication</summary>

```
vu 2, record 2
vu 7, record 7
vu 4, record 7
vu 1, record 7
```
Le record passe à 7 au deuxième tour, puis 4 et 1 ne le battent pas : il reste à 7. Le `print` est au niveau du
`if` (4 espaces), donc dans la boucle mais pas dans le `if` : il s'exécute à chaque tour.

</details>

### La série : un compteur qui repart à zéro

Quelle est la plus longue suite de jours de soleil d'affilée ? Il faut deux variables :

- `serie` : la longueur de la série **en cours**. Un jour de soleil l'allonge de 1 ; un autre jour la **remet à 0**
  (la série est cassée).
- `meilleure_serie` : le **record** des séries, mis à jour après chaque jour, comme le record ci-dessus.

````python
ciel = "SPSSSPS"
serie = 0
meilleure_serie = 0
for jour in ciel:
    if jour == "S":
        serie = serie + 1           # la série continue
    else:
        serie = 0                   # la série est cassée
    if serie > meilleure_serie:     # au niveau du for : testé à chaque tour
        meilleure_serie = serie
print(f"Plus longue série de soleil : {meilleure_serie} jours")
````

| Tour | `jour` | `serie` après | `meilleure_serie` après |
|---|---|---|---|
| 1 | S | 1 | 1 |
| 2 | P | 0 | 1 |
| 3 | S | 1 | 1 |
| 4 | S | 2 | 2 |
| 5 | S | 3 | 3 |
| 6 | P | 0 | 3 |
| 7 | S | 1 | 3 |

Regarde l'indentation : le deuxième `if` est au même niveau que le premier (4 espaces), **après** le
`if / else`. Il est donc exécuté à chaque tour, quelle que soit la branche prise.

🔮 **Prédis** : Avant d'exécuter : que va afficher cette cellule ? Écris ta réponse sur papier (ou dans ta tête), puis exécute.

````python
serie = 0
meilleure_serie = 0
for jour in "SSPSPP":
    if jour == "S":
        serie = serie + 1
    else:
        serie = 0
    if serie > meilleure_serie:
        meilleure_serie = serie
print(serie, meilleure_serie)
````

<details><summary>👀 Voir l'explication</summary>

`0 2`. La meilleure série fait 2 (les deux premiers jours). À la fin, `serie` vaut 0 : le dernier jour est un `P`,
il a cassé la série en cours. Le record, lui, est resté à 2.

</details>

## 6. Accumuler du texte

L'accumulateur marche aussi avec du texte. On part d'une **chaîne vide** `""` (deux guillemets collés, zéro
caractère) et on **colle** un morceau à chaque tour avec `+` (la concaténation du module 02).

````python
barre = ""
for i in range(5):
    barre = barre + "#"
print(barre)
````

Cinq tours, cinq `#` collés : `#####`. Avec ça, on peut dessiner un petit graphique en texte.

On peut aussi transformer chaque lettre et reconstruire un nouveau texte :

````python
sigle = ""
for lettre in "meteo":
    sigle = sigle + lettre.upper() + "."
print(sigle)
````

🔮 **Prédis** : Avant d'exécuter : que va afficher cette cellule ? Écris ta réponse sur papier (ou dans ta tête), puis exécute.

````python
mot = ""
for lettre in "abc":
    mot = lettre + mot
print(mot)
````

<details><summary>👀 Voir l'explication</summary>

`cba`. Ici on colle la nouvelle lettre **devant** (`lettre + mot`) et non derrière. Tour 1 : `"a"`. Tour 2 :
`"b" + "a"` = `"ba"`. Tour 3 : `"c" + "ba"` = `"cba"`. Le mot est retourné.

</details>

### Lire des nombres cachés dans une chaîne

Le capteur de MétéoLocale envoie parfois plusieurs nombres dans un seul texte, séparés par des points-virgules :
`"12.0;13.5;11.8;"`. Chaque nombre a plusieurs caractères : on ne peut plus faire `int(chiffre)` caractère par
caractère.

L'astuce : un **tampon**, une chaîne qui sert de brouillon. On y colle les caractères un par un ; quand on
rencontre un `;`, le nombre est complet : on le convertit, on l'ajoute au total, et on vide le tampon.

````python
mesures = "12.0;13.5;11.8;"
tampon = ""
total = 0
for caractere in mesures:
    if caractere == ";":
        total = total + float(tampon)   # le nombre est complet : on l'ajoute
        tampon = ""                     # on vide le brouillon
    else:
        tampon = tampon + caractere     # on continue de recopier le nombre
print(f"Total : {total:.1f}")
````

| Caractère lu | Que fait-on ? | `tampon` après | `total` après |
|---|---|---|---|
| `1`, `2`, `.`, `0` | on colle | `"12.0"` | 0 |
| `;` | on ajoute `float("12.0")`, on vide | `""` | 12.0 |
| `1`, `3`, `.`, `5` | on colle | `"13.5"` | 12.0 |
| `;` | on ajoute 13.5, on vide | `""` | 25.5 |
| `1`, `1`, `.`, `8` | on colle | `"11.8"` | 25.5 |
| `;` | on ajoute 11.8, on vide | `""` | 37.3 |

Le `;` final est important : sans lui, le dernier nombre resterait dans le tampon. Au module 05, tu verras
`.split(";")` découper tout ça en une ligne. Mais tu sauras ce qui se passe à l'intérieur.

## 7. La boucle `while` : tant que…

`for` sert quand on sait **sur quoi** on boucle (une chaîne, un `range`). Parfois, on ne sait pas combien de
tours il faudra : on veut continuer **tant qu'**une condition est vraie. C'est la boucle `while`
(« tant que » en anglais).

Exemple : le matin, il fait 12 °C et la température monte de 1.5 °C par heure. Au bout de combien d'heures
dépasse-t-on 18 °C ?

````python
temperature = 12.0
heures = 0
while temperature <= 18:
    temperature = temperature + 1.5
    heures = heures + 1
print(f"Après {heures} h : {temperature} °C")
````

Comment Python lit un `while` :

1. Il teste la condition (`temperature <= 18`, une comparaison du module 03).
2. Si elle est **vraie** (`True`), il exécute le corps indenté, puis **remonte** au test.
3. Dès qu'elle est **fausse** (`False`), il saute le corps et passe à la suite.

| Test | `temperature` | condition | on fait un tour ? |
|---|---|---|---|
| 1 | 12.0 | `12.0 <= 18` vrai | oui → 13.5, heures = 1 |
| 2 | 13.5 | vrai | oui → 15.0, heures = 2 |
| 3 | 15.0 | vrai | oui → 16.5, heures = 3 |
| 4 | 16.5 | vrai | oui → 18.0, heures = 4 |
| 5 | 18.0 | `18.0 <= 18` vrai | oui → 19.5, heures = 5 |
| 6 | 19.5 | faux | non : fin |

🔮 **Prédis** : Avant d'exécuter : que va afficher cette cellule ? Écris ta réponse sur papier (ou dans ta tête), puis exécute.

````python
reserve = 50
jours = 0
while reserve > 0:
    reserve = reserve - 20
    jours = jours + 1
print(f"{jours} jours, il reste {reserve}")
````

<details><summary>👀 Voir l'explication</summary>

`3 jours, il reste -10`. 50 → 30 → 10 → -10. Après le 3e tour, `reserve` vaut -10 : `-10 > 0` est faux, la
boucle s'arrête. Le test n'a lieu qu'**au début** de chaque tour : la réserve peut passer sous zéro pendant le
dernier tour.

</details>

### ⚠️ Le grand danger : la boucle infinie

Si la condition du `while` ne devient **jamais** fausse, la boucle ne s'arrête jamais : c'est une **boucle
infinie**. Exemple (⛔ ne le tape pas, il est ici en simple texte) :

```python
temperature = 12.0
while temperature <= 18:
    print("Il fait encore frais")
    # oubli : temperature ne change jamais !
```

`temperature` reste à 12.0 pour toujours, donc `12.0 <= 18` est toujours vrai. Le notebook affiche des lignes
sans fin et la cellule garde son petit indicateur « en cours » (`[*]` ou un rond qui tourne).

**Comment t'en sortir** : clique sur le bouton carré **■ Interrupt** (« interrompre ») en haut du notebook.
Dans VS Code, il se trouve dans la barre au-dessus des cellules ; dans Jupyter, dans la barre d'outils (ou menu
*Kernel → Interrupt*). Python s'arrête et affiche `KeyboardInterrupt` : ce n'est pas grave, c'est juste
« arrêté à la main ».

**Comment l'éviter** : dans le corps d'un `while`, vérifie qu'**une ligne change la variable testée** dans la
condition, et dans le bon sens (si on attend qu'elle dépasse 18, elle doit augmenter).

Une ceinture de sécurité possible : un compteur de tours qui coupe la boucle au-delà d'un maximum.

````python
temperature = 12.0
tours = 0
while temperature <= 18 and tours < 100:     # sécurité : jamais plus de 100 tours
    temperature = temperature + 1.5
    tours = tours + 1
print(f"{tours} tours")
````

## 8. `break` et `continue`

### `break` : sortir tout de suite

`break` (« casser ») arrête la boucle immédiatement, même s'il reste des tours. On s'en sert pour **chercher** :
dès qu'on a trouvé, inutile de continuer. Exemple : quel est le premier jour de pluie de la semaine ?

````python
ciel = "SSNPSPS"
numero = 0
premier_jour_pluie = 0
for jour in ciel:
    numero = numero + 1
    if jour == "P":
        premier_jour_pluie = numero
        break                     # trouvé : on sort de la boucle
print(f"Première pluie le jour {premier_jour_pluie}")
````

`numero` est un compteur de tours qui donne le numéro du jour (1, 2, 3…). Au 4e tour, on tombe sur `P` : on
range 4 et `break` nous fait sortir. Le `P` du 6e jour n'est jamais lu.

🔮 **Prédis** : Avant d'exécuter : que va afficher cette cellule ? Écris ta réponse sur papier (ou dans ta tête), puis exécute.

````python
for i in range(1, 10):
    if i == 4:
        break
    print(i)
print("sorti")
````

<details><summary>👀 Voir l'explication</summary>

`1`, `2`, `3`, puis `sorti`. Au tour où `i` vaut 4, `break` sort de la boucle **avant** le `print(i)` : le 4
n'est jamais affiché, ni les nombres suivants.

</details>

### `continue` : passer au tour suivant

`continue` saute **la fin du tour en cours** et passe directement au tour suivant. On s'en sert pour
**ignorer** certaines valeurs. Exemple : le capteur écrit `?` les jours où il était en panne.

````python
pluies = "3?12?0"
total = 0
nb_mesures = 0
for caractere in pluies:
    if caractere == "?":
        continue                  # mesure manquante : on passe au caractère suivant
    total = total + int(caractere)
    nb_mesures = nb_mesures + 1
print(f"{nb_mesures} mesures, total {total} mm, moyenne {total / nb_mesures:.1f} mm")
````

Sans le `continue`, `int("?")` planterait avec une `ValueError` (module 02). Ici le compteur `nb_mesures` est
indispensable : il ne compte que les vraies mesures (4), pas les 6 caractères.

🔮 **Prédis** : Avant d'exécuter : que va afficher cette cellule ? Écris ta réponse sur papier (ou dans ta tête), puis exécute.

````python
resultat = ""
for lettre in "a-b-c":
    if lettre == "-":
        continue
    resultat = resultat + lettre
print(resultat)
````

<details><summary>👀 Voir l'explication</summary>

`abc`. Les tirets sont ignorés : pour eux, `continue` saute la ligne qui colle la lettre.

</details>

**Résumé** : `break` = « j'arrête tout, je sors » ; `continue` = « je passe ce tour, je continue la boucle ».
Les deux s'écrivent seuls sur leur ligne, presque toujours dans un `if`. Ils marchent dans `for` comme dans
`while`.

### ⚠️ Erreur fréquente : `break` hors d'une boucle

````python
pluie = 12
if pluie > 10:
    break
````

1. Dernière ligne : `SyntaxError: 'break' outside loop` : « `break` en dehors d'une boucle ».
2. Un `if` n'est **pas** une boucle : il n'y a rien à « casser ».

✅ Correction : `break` et `continue` ne s'utilisent qu'à l'intérieur d'un `for` ou d'un `while`. Si tu veux juste
ne rien faire de plus, écris le reste du code dans un `else`.
