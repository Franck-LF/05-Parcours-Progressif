## Comment travailler

1. Lis l'**objectif** et les étapes.
2. Remplace chaque `...` par ton code ; **garde les noms de variables déjà écrits**, la vérification en a besoin.
3. Exécute ta cellule (`Shift + Entrée`) et compare au **résultat attendu**.
4. Exécute la cellule **Vérification** juste en dessous : `✅ OK` = réussi ; sinon lis le message, il te dit
   quoi corriger.
5. Bloqué plus de 5 minutes ? Ouvre l'indice 💡, puis relis la section du cours indiquée.

> 🩺 **Lire un échec de vérification**
> - Ne lis que la **dernière ligne** du message.
> - `AssertionError: <message>` veut dire « la vérification a échoué » : le message **en français** après les
>   deux-points te dit quoi corriger. Le reste (en anglais, au-dessus) peut être ignoré.
> - Si la dernière ligne commence par un autre mot (`NameError`, `TypeError`, `SyntaxError`…), c'est **ta cellule**
>   qui a un problème : relis-la, puis ré-exécute-la avant la vérification.
> - Les caractères spéciaux (°, →, ☀, ☂…) se copient-collent depuis l'énoncé.

💬 Si un message parle de `Ellipsis` ou de `'ellipsis' object`, c'est qu'il reste un `...` à remplacer dans ta
cellule.

🔁 Une boucle qui ne s'arrête pas ? Bouton carré **■ Interrupt** en haut du notebook (cours, section 7).

Les trois niveaux : **Imiter** (le geste du cours, presque à l'identique), **Adapter** (le même geste dans un
autre contexte), **Transposer** (un petit problème à résoudre seul). Le même geste revient plusieurs fois : c'est
voulu, c'est comme ça qu'il s'ancre.
