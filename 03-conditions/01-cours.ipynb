## 1. Comparer : la réponse est `True` ou `False`

Une **comparaison** pose une question à Python, et il répond par un **booléen** (le type `bool` du module 01) :
`True` (vrai) ou `False` (faux). C'est comme dans Excel, où la formule `=B2>25` affiche `VRAI` ou `FAUX`.

| Symbole | Question posée | Exemple | Résultat |
|---|---|---|---|
| `==` | est égal à ? (**deux** signes égal) | `11 == 11` | `True` |
| `!=` | est différent de ? | `11 != 41` | `True` |
| `<` | est strictement plus petit ? | `12.0 < 10` | `False` |
| `>` | est strictement plus grand ? | `25.9 > 25` | `True` |
| `<=` | est plus petit **ou égal** ? | `28 <= 28` | `True` |
| `>=` | est plus grand **ou égal** ? | `27.9 >= 28` | `False` |

````python
temp_max = 25.9
print(temp_max > 25)
print(temp_max >= 28)
````

Bon à savoir : un `float` et un `int` qui ont la même valeur sont **égaux** pour Python (`12.0 == 12` est vrai).

🔮 **Prédis** : Avant d'exécuter : que va afficher cette cellule ? Écris ta réponse sur papier (ou dans ta tête), puis exécute.

````python
pluie_mm = 0.0
print(pluie_mm == 0)
print(pluie_mm != 0)
````

<details><summary>👀 Voir l'explication</summary>

```
True
False
```
`0.0 == 0` est vrai : pour Python, `0.0` et `0` sont le même nombre, même si l'un est un `float` et l'autre un
`int`. Donc « est-ce différent de 0 ? » est faux.

</details>

Le résultat d'une comparaison est une valeur comme une autre : on peut la **ranger dans une variable**. Python
calcule d'abord la comparaison, à droite du `=`, puis range `True` ou `False` à gauche.

````python
temp_max = 29.2
chaleur = temp_max >= 28
print(chaleur, type(chaleur))
````

`False`. `>` veut dire **strictement** plus grand : 40 n'est pas plus grand que 40. Pour « 40 ou plus », il faut
`>=`. C'est le piège classique des seuils : relis toujours l'énoncé (« au moins 40 » = `>= 40`).

</details>

### `=` et `==` : ne pas les confondre

| Écriture | Sens | Exemple |
|---|---|---|
| `=` (un signe) | **ranger** une valeur dans une variable | `ville = "Albi"` |
| `==` (deux signes) | **comparer** deux valeurs | `ville == "Albi"` → `True` ou `False` |

Moyen mnémotechnique : `==` pose une question, il a besoin de deux signes pour hésiter.

## 3. `if` : faire quelque chose seulement si…

Le mot **`if`** (« si » en anglais) permet d'exécuter des lignes **seulement si** une condition est vraie. C'est
la fonction `SI()` d'Excel, ou la consigne d'Inès : « **si** il a plu, ajoute un parapluie au bulletin ».

````python
pluie_mm = 17.4
if pluie_mm > 0:
    print("Prends un parapluie !")
print("Fin du bulletin")
````

**Lis-la attentivement, c'est la forme de toutes les conditions** :

1. `if`, puis la **condition** (une comparaison), puis **deux-points `:`** à la fin de la ligne.
2. Les lignes à exécuter « si c'est vrai » sont **décalées vers la droite de 4 espaces**. Ce décalage s'appelle
   l'**indentation**, et l'ensemble des lignes décalées s'appelle un **bloc**. En Python, l'indentation n'est
   pas décorative : **c'est elle qui dit où le bloc commence et où il finit**.
3. La première ligne **qui n'est plus décalée** (`print("Fin du bulletin")`) ne fait plus partie du bloc : elle
   s'exécute **toujours**.

Dans VS Code, après les deux-points, la touche **Entrée** décale souvent la ligne suivante toute seule. Sinon,
la touche **Tab** ajoute 4 espaces, et **Maj + Tab** les retire.

🔮 **Prédis** : Avant d'exécuter : que va afficher cette cellule ? Écris ta réponse sur papier (ou dans ta tête), puis exécute.

````python
pluie_mm = 0.0
if pluie_mm > 0:
    print("Prends un parapluie !")
    print("Il pleut à Toulouse")
print("Fin du bulletin")
````

<details><summary>👀 Voir l'explication</summary>

Seulement `Fin du bulletin`. La condition `0.0 > 0` est fausse : **tout le bloc** (les deux lignes décalées) est
sauté. La dernière ligne n'est pas décalée : elle s'exécute dans tous les cas.

</details>

## 4. `else` : sinon…

**`else`** (« sinon ») donne le bloc à exécuter quand la condition est **fausse**. Un des deux blocs s'exécute
toujours, jamais les deux. `else` n'a **pas** de condition, mais il a ses deux-points, et il est aligné avec son
`if` (pas décalé).

````python
temp_max = 25.9
if temp_max >= 28:
    etat = "chaud"
else:
    etat = "pas chaud"
print(etat)
````

Remarque le geste : plutôt que d'afficher dans chaque bloc, on **range** le résultat dans une variable (`etat`),
puis on l'affiche (ou on l'utilise) **après** le `if`. C'est ce qu'on fera presque tout le temps.

🔮 **Prédis** : Avant d'exécuter : que va afficher cette cellule ? Écris ta réponse sur papier (ou dans ta tête), puis exécute.

````python
vent_kmh = 52
if vent_kmh >= 40:
    conseil = "Rentre les parasols"
else:
    conseil = "Rien à signaler"
print(conseil)
````

<details><summary>👀 Voir l'explication</summary>

`Rentre les parasols`. `52 >= 40` est vrai : le premier bloc s'exécute, le bloc du `else` est sauté.

</details>

## 5. `elif` : plusieurs cas

Pour plus de deux cas, on ajoute des **`elif`** (contraction de *else if*, « sinon, si »). Python teste les
conditions **dans l'ordre, de haut en bas**, et exécute le bloc de la **première** condition vraie. Ensuite, il
saute tout le reste, même si d'autres conditions plus bas étaient vraies aussi. Le `else` final (facultatif)
attrape tous les cas restants.

````python
pluie_mm = 17.4
if pluie_mm >= 10:
    classe = "forte pluie"
elif pluie_mm >= 1:
    classe = "pluie"
else:
    classe = "sec"
print(classe)
````

Python lit : « 17.4 ≥ 10 ? oui → `forte pluie`, et j'arrête ». Avec `pluie_mm = 3`, il lirait : « 3 ≥ 10 ? non.
3 ≥ 1 ? oui → `pluie`, et j'arrête ». Avec `0.0` : non, non → `else` → `sec`.

🔮 **Prédis** : Avant d'exécuter : que va afficher cette cellule ? Écris ta réponse sur papier (ou dans ta tête), puis exécute.

````python
pluie_mm = 17.4
if pluie_mm >= 1:
    classe = "pluie"
elif pluie_mm >= 10:
    classe = "forte pluie"
else:
    classe = "sec"
print(classe)
````

<details><summary>👀 Voir l'explication</summary>

`pluie`, et non `forte pluie` ! La première condition `17.4 >= 1` est déjà vraie : Python s'arrête là et ne
regarde même pas le `elif`. Avec cet ordre, la branche `forte pluie` ne peut **jamais** être atteinte.

**Règle** : avec des seuils, mets **la condition la plus exigeante en premier** (`>= 10` avant `>= 1`).

</details>

🔮 **Prédis** : Avant d'exécuter : que va afficher cette cellule ? Écris ta réponse sur papier (ou dans ta tête), puis exécute.

````python
temp_max = 15.0
if temp_max < 15:
    ressenti = "frais"
elif temp_max < 25:
    ressenti = "doux"
else:
    ressenti = "chaud"
print(ressenti)
````

<details><summary>👀 Voir l'explication</summary>

`doux`. `15.0 < 15` est faux (15 n'est pas **strictement** plus petit que 15). Puis `15.0 < 25` est vrai →
`doux`. Les seuils exacts sont toujours les cas à vérifier.

</details>

## 6. Combiner des conditions : `and`, `or`, `not`

| Mot | Sens | Vrai quand… | Exemple |
|---|---|---|---|
| `and` | et | les **deux** conditions sont vraies | `pluie_mm == 0 and vent_kmh < 20` (beau temps calme) |
| `or` | ou | **au moins une** des deux est vraie | `pluie_mm >= 10 or vent_kmh >= 50` (alerte) |
| `not` | non | la condition est **fausse** | `not il_a_plu` |

Avec `and`, une seule condition fausse suffit à tout rendre faux. Avec `or`, une seule condition vraie suffit à
tout rendre vrai.

````python
pluie_mm = 0.0
vent_kmh = 11
beau_temps = pluie_mm == 0 and vent_kmh < 20
print(beau_temps)
````

🔮 **Prédis** : Avant d'exécuter : que va afficher cette cellule ? Écris ta réponse sur papier (ou dans ta tête), puis exécute.

````python
pluie_mm = 3.5
vent_kmh = 53
print(pluie_mm >= 10 and vent_kmh >= 50)
print(pluie_mm >= 10 or vent_kmh >= 50)
````

<details><summary>👀 Voir l'explication</summary>

```
False
True
```
`3.5 >= 10` est faux, `53 >= 50` est vrai. Avec `and`, il faut les deux : faux. Avec `or`, une seule suffit :
vrai.

</details>

🔮 **Prédis** : Avant d'exécuter : que va afficher cette cellule ? Écris ta réponse sur papier (ou dans ta tête), puis exécute.

````python
il_a_plu = False
print(not il_a_plu)
````

<details><summary>👀 Voir l'explication</summary>

`True`. `not` inverse : « pas faux », c'est vrai. On s'en sert pour lire naturellement :
`if not il_a_plu:` = « s'il n'a pas plu ».

</details>

**Un booléen suffit comme condition.** Si une variable contient déjà `True` ou `False`, on peut l'écrire
directement après `if` : `if chaleur:` se lit « s'il fait chaud » et veut dire exactement `if chaleur == True:`.

````python
temp_max = 29.2
chaleur = temp_max >= 28
if chaleur:
    conseil = "Bois de l'eau"
else:
    conseil = "Rien de spécial"
print(conseil)
````

⚠️ Chaque côté de `and` / `or` doit être une **comparaison complète**. On n'écrit pas `vent_kmh >= 40 and < 60`,
mais `vent_kmh >= 40 and vent_kmh < 60`.

### `and` et `or` ensemble : les parenthèses

Quand une condition mélange `and` et `or`, Python fait **d'abord les `and`**, puis les `or` (comme `*` passe
avant `+`). Pour ne pas avoir à retenir cette règle, et pour que le code soit lisible, **mets des parenthèses**
autour de ce qui va ensemble : comme en calcul, ce qui est entre parenthèses est évalué d'abord, d'un bloc.

Exemple : canicule si la maximale atteint 35 °C, **ou** si la nuit est chaude (minimale ≥ 20) **et** la journée
aussi (maximale ≥ 30) :

```python
canicule = temp_max >= 35 or (temp_min >= 20 and temp_max >= 30)
```

🔮 **Prédis** : Avant d'exécuter : que va afficher cette cellule ? Écris ta réponse sur papier (ou dans ta tête), puis exécute.

````python
pluie_mm = 0.0
vent_kmh = 60
temp_max = 15.0
print((pluie_mm >= 1 or vent_kmh >= 40) and temp_max >= 28)
print(pluie_mm >= 1 or (vent_kmh >= 40 and temp_max >= 28))
````

<details><summary>👀 Voir l'explication</summary>

```
False
False
```
Ligne 1 : la parenthèse donne `False or True` = `True`, puis `True and False` = `False`.
Ligne 2 : la parenthèse donne `True and False` = `False`, puis `False or False` = `False`.

Change `pluie_mm` en `3.0` et ré-exécute : la ligne 1 reste `False` (il ne fait pas chaud), mais la ligne 2 devient
`True` (il pleut, et ça suffit avec `or`). **Les parenthèses changent le sens** : place-les là où ta phrase en
français met « et » ou « ou » ensemble.

</details>

Une condition avec `and`/`or` s'utilise comme les autres dans un `if` :

````python
pluie_mm = 0.0
vent_kmh = 11
if pluie_mm == 0 and vent_kmh < 20:
    conseil = "Parfait pour un pique-nique"
else:
    conseil = "Reporte le pique-nique"
print(conseil)
````

## 7. Une condition dans une condition

Un bloc peut contenir un autre `if` : c'est une condition **imbriquée**. Le deuxième bloc est décalé **encore** de
4 espaces, soit **8 espaces** au total. Lis l'indentation comme les niveaux d'un sommaire.

````python
pluie_mm = 14.0
vent_kmh = 41
if pluie_mm >= 1:
    if vent_kmh >= 40:
        temps = "tempête"
    else:
        temps = "pluie"
else:
    temps = "sec"
print(temps)
````

🔮 **Prédis** : Avant d'exécuter : que va afficher cette cellule ? Écris ta réponse sur papier (ou dans ta tête), puis exécute.

````python
pluie_mm = 0.0
vent_kmh = 60
if pluie_mm >= 1:
    if vent_kmh >= 40:
        temps = "tempête"
    else:
        temps = "pluie"
else:
    temps = "sec"
print(temps)
````

<details><summary>👀 Voir l'explication</summary>

`sec`. La première condition `0.0 >= 1` est fausse : Python saute **tout** le bloc du premier `if`, y compris le
`if` imbriqué. Le vent n'est même pas regardé. On va directement au `else` du premier niveau.

</details>

Souvent, on peut écrire la même chose sans imbriquer, avec `and` et `elif`. Les deux sont corrects ; choisis le
plus lisible :

```python
if pluie_mm >= 1 and vent_kmh >= 40:
    temps = "tempête"
elif pluie_mm >= 1:
    temps = "pluie"
else:
    temps = "sec"
```

## 8. ⚠️ Erreur fréquente : comparer un texte et un nombre

Une valeur lue dans une ligne brute (module 02) est un **texte**. Si on la compare à un nombre, Python refuse :

````python
morceaux = "Toulouse,25.9".split(",")
chaud = morceaux[1] > 25
````

**Lire le message** : `TypeError: '>' not supported between instances of 'str' and 'int'` = « `>` n'est pas
possible entre un `str` et un `int` ». On ne peut pas dire si un texte est « plus grand » qu'un nombre.

**Correction** : convertir d'abord avec `float()` (module 02) :

````python
morceaux = "Toulouse,25.9".split(",")
chaud = float(morceaux[1]) > 25
print(chaud)
````
