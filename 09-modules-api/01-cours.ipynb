## 1. `import` : ouvrir un tiroir de la boîte à outils

Un **module**, c'est simplement **un fichier Python** (`.py`) rempli de fonctions et de variables prêtes à
l'emploi, écrites par quelqu'un d'autre (ou par toi !). Python est livré avec des centaines de modules : c'est la
**bibliothèque standard**, une grande boîte à outils dont chaque module est un **tiroir**.

`import math` ouvre le tiroir `math` (les mathématiques). Ensuite, on prend un outil avec un **point** :
`math.sqrt` se lit « l'outil `sqrt` du tiroir `math` » (*square root* = racine carrée).

````python
import math

print(math.sqrt(16))
print(math.pi)
````

Pourquoi des tiroirs, et pas tout en vrac ? Pour ne pas tout charger d'un coup, et pour éviter les conflits de
noms : `math.floor` et un éventuel `floor` à toi ne se marchent pas dessus.

Et la « formule magique » du module 01 ? `import sys` ouvre le tiroir `sys` (le *système* : des informations sur
Python lui-même), et `sys.executable` est une variable de ce tiroir : le chemin du Python qui exécute ce notebook.
Tu comprends maintenant chaque mot.

````python
import sys

print(sys.executable)
````

Si le chemin affiché contient `.venv`, ton notebook tourne bien dans l'environnement virtuel de ton projet
(module 00). On en reparle à la section 7.

## 2. Les trois façons d'importer

| Forme | Exemple | Ensuite on écrit |
|---|---|---|
| Tout le tiroir | `import math` | `math.sqrt(16)` |
| Un ou plusieurs outils | `from math import sqrt, pi` | `sqrt(16)`, `pi` |
| Tout le tiroir, avec un surnom | `import statistics as stats` | `stats.mean([...])` |

- `import math` est la forme la plus claire : on voit d'où vient chaque outil.
- `from ... import ...` est pratique quand on utilise souvent le même outil (comme `from pathlib import Path`).
- `as` donne un **surnom** (un alias) à un nom long. Tu le verras partout au module 10 : `import pandas as pd`.

````python
from math import sqrt, pi
import statistics as stats

print(sqrt(25))
print(round(pi, 2))
print(stats.mean([25.9, 23.2, 27.2]))
````

## 3. Le module `random` : le hasard... contrôlé

`random` (*aléatoire*) tire des nombres au hasard : utile pour simuler, mélanger, tirer un échantillon.

- `random.randint(1, 6)` : un entier entre 1 et 6 (inclus), comme un dé ;
- `random.choice(liste)` : un élément de la liste au hasard ;
- `random.uniform(-1, 1)` : un nombre décimal entre -1 et 1.

En data, on veut pouvoir **refaire exactement** la même expérience. `random.seed(42)` fixe la **graine** du
hasard : avec la même graine, on obtient **toujours la même suite** de tirages. Analogie : mélanger un jeu de cartes
toujours de la même façon, en partant de la même position.

````python
import random

villes = ["Toulouse", "Montpellier", "Nîmes", "Perpignan", "Albi", "Tarbes"]

random.seed(42)
print(random.choice(villes), random.randint(1, 6))

random.seed(42)
print(random.choice(villes), random.randint(1, 6))
````

<details><summary>👀 Voir l'explication</summary>

`[2, 5, 1, 3, 1]` puis `2`. On ne peut pas deviner les valeurs (c'est du hasard), mais on peut prédire que le
dernier tirage vaut **le premier de la liste** : même graine (`1`), même départ. Sans `seed`, chaque exécution
donnerait autre chose. Dans les exercices, on fixe toujours la graine pour que la vérification marche.

</details>

## 4. Le module `datetime` : des vraies dates

Jusqu'ici nos dates étaient du texte, `"2026-09-01"`. Avec du texte, impossible de demander « quel jour de la
semaine ? » ou « la date 7 jours plus tard ? ». Le module `datetime` fournit deux outils :

- `date` : une date (année, mois, jour) ;
- `timedelta` : une **durée** (*delta* = écart), par exemple 7 jours.

````python
from datetime import date, timedelta

premier = date(2026, 9, 1)
print(premier)
print(premier.year, premier.month, premier.day)

octobre = date.fromisoformat("2026-10-01")     # depuis le texte du fichier
print(octobre + timedelta(days=7))
print((octobre - premier).days, "jours d'écart")
````

- `date.fromisoformat("2026-10-01")` lit le format **ISO** (année-mois-jour), celui de nos fichiers ;
- une date + un `timedelta` donne une autre date ; une date − une date donne un `timedelta`, dont `.days` est le
  nombre de jours ;
- `.isoformat()` fait l'inverse de `fromisoformat` : `date(2026, 9, 1).isoformat()` → `"2026-09-01"`.

`.weekday()` donne le **jour de la semaine** sous forme de nombre : **0 = lundi**, 1 = mardi, ..., 6 = dimanche.
Pour avoir le nom en français, on utilise ce nombre comme indice dans une liste :

````python
JOURS = ["lundi", "mardi", "mercredi", "jeudi", "vendredi", "samedi", "dimanche"]

print(premier.weekday(), JOURS[premier.weekday()])
````

`2026-10-01 jeudi`. Le 30 septembre + 1 jour = le **1ᵉʳ octobre** : `timedelta` gère tout seul les fins de mois
(et les années bissextiles). Le 30 septembre 2026 est un mercredi (`weekday()` = 2), donc le lendemain est un jeudi.

</details>

## 5. Le module `statistics` : les statistiques de base

Tu as écrit ta propre fonction `moyenne` au module 07. Le module `statistics` en fournit une, et d'autres :

| Outil | Ce qu'il calcule |
|---|---|
| `statistics.mean(valeurs)` | la **moyenne** |
| `statistics.median(valeurs)` | la **médiane** : la valeur du milieu quand on trie (la moitié des valeurs est en dessous) |
| `statistics.stdev(valeurs)` | l'**écart-type** : de combien, en gros, les valeurs s'écartent de la moyenne |

La médiane résiste mieux aux valeurs extrêmes : un seul jour à 40 °C fait monter la moyenne, presque pas la médiane.

````python
import statistics

temps = [25.9, 23.2, 27.2, 26.1, 25.3]
print(statistics.mean(temps))
print(statistics.median(temps))
print(round(statistics.stdev(temps), 2))
````

En data, on regarde **les deux** : un gros écart entre moyenne et médiane signale des valeurs extrêmes.

### Le piège du cache : modifier un module déjà importé

Python **garde en mémoire** les modules déjà importés (c'est un **cache**, une mémoire de travail) : un deuxième
`import` du même module ne relit **pas** le fichier. C'est rapide... mais si tu **modifies** le fichier, Python
continue d'utiliser l'ancienne version !

````python
code_v2 = """
def moyenne(valeurs):
    '''Version 2 : moyenne ARRONDIE à 1 chiffre.'''
    return round(sum(valeurs) / len(valeurs), 1)

def celsius_vers_fahrenheit(celsius, chiffres=1):
    return round(celsius * 9 / 5 + 32, chiffres)

VILLES = ["Toulouse", "Montpellier", "Nîmes", "Perpignan", "Albi", "Tarbes"]
"""
with open("sorties/meteo_outils.py", "w", encoding="utf-8") as fichier:
    fichier.write(code_v2)

from sorties import meteo_outils
print(meteo_outils.moyenne([25.9, 23.2, 27.2]))     # toujours l'ancienne version !
````

Deux solutions simples :
1. **Recharger** le module avec `importlib.reload(...)` (le module `importlib` sert à piloter les imports) ;
2. ou **redémarrer le kernel** (menu *Kernel → Restart*) : la mémoire est vidée, le prochain `import` relit tout.

Dans les exercices, chaque exercice qui crée un module lui donne **un nom différent** (`outils_14`, `outils_26`...),
pour qu'il n'y ait jamais d'ancienne version en mémoire, et ajoute une ligne `importlib.reload(...)` de sécurité :
si tu corriges ton fichier et relances la cellule, c'est bien la nouvelle version qui est utilisée.

````python
import importlib

importlib.reload(meteo_outils)
print(meteo_outils.moyenne([25.9, 23.2, 27.2]))     # la version 2, arrondie
````

## 8. Une API HTTP : poser une question à un site, recevoir du JSON

Un site web comme open-meteo.com a deux portes :
- une porte pour les **humains** : des pages jolies, avec des images ;
- une porte pour les **programmes** : l'**API** (*Application Programming Interface*, interface de programmation).

Analogie : l'API est un **guichet**. Tu tends un formulaire rempli (l'**adresse** de la requête, avec des
**paramètres**), le guichetier te rend une fiche (la **réponse**), écrite en **JSON** (le format du module 08).

**HTTP** est la langue des échanges sur le web : ton navigateur l'utilise à chaque page. Une requête HTTP qui
demande des données s'appelle un **GET** (« obtenir »).

Voici l'adresse (l'**URL**) qui demande à Open-Meteo les prévisions de Toulouse :

```
https://api.open-meteo.com/v1/forecast?latitude=43.6&longitude=1.44&daily=temperature_2m_max,temperature_2m_min,precipitation_sum&timezone=Europe/Paris
```

- avant le `?` : le guichet (`/v1/forecast`, les prévisions) ;
- après le `?` : les **paramètres**, sous la forme `nom=valeur`, séparés par des `&` : la position (latitude,
  longitude), les données quotidiennes (`daily`) voulues, le fuseau horaire.

Tu peux coller cette adresse dans ton navigateur : il affiche le JSON brut de la réponse.

## 9. `requests` : faire la requête depuis Python

Le paquet `requests` (installé avec `requirements.txt`) envoie la requête pour toi :
- `requests.get(url, params=parametres, timeout=10)` : envoie le GET ; `params` est un **dictionnaire** des
  paramètres (requests fabrique la partie après le `?`) ; `timeout=10` : abandonner après 10 secondes sans réponse ;
- `reponse.status_code` : le **code de statut** HTTP, un nombre : `200` = OK, `404` = page introuvable,
  `500` = le serveur a un problème ;
- `reponse.json()` : transforme le JSON de la réponse en dictionnaires et listes Python (comme `json.load`).

Une requête réseau peut échouer de **beaucoup** de façons (pas de Wi-Fi, site en panne, réseau d'entreprise
qui bloque, `requests` pas installé...). C'est le cas où l'on accepte d'attraper **toute** erreur : `except
Exception as erreur` (`Exception` = la famille de toutes les erreurs courantes ; `as erreur` range l'erreur dans
une variable pour l'afficher). On ne cache rien : on dit ce qui s'est passé, et on passe au **plan B**, le fichier
figé.

⚠️ **À savoir avant d'exécuter** : avec du réseau, l'API renvoie les prévisions **d'aujourd'hui**, donc des dates
**différentes** de celles du fichier figé (1ᵉʳ au 7 octobre 2026) utilisé dans la suite du cours et les exercices.
C'est normal : tu n'as rien cassé. Sans réseau, tu verras les dates du fichier figé.

````python
import json

URL = "https://api.open-meteo.com/v1/forecast"
parametres = {
    "latitude": 43.6,
    "longitude": 1.44,
    "daily": "temperature_2m_max,temperature_2m_min,precipitation_sum",
    "timezone": "Europe/Paris",
}

try:
    import requests
    reponse = requests.get(URL, params=parametres, timeout=10)
    print("Code de statut :", reponse.status_code)
    en_direct = reponse.json()
    jours_recus = en_direct["daily"]["time"]
except Exception as erreur:
    print("Pas de réponse utilisable de l'API (pas de réseau ?) :", type(erreur).__name__)
    print("Plan B : on lit la réponse enregistrée dans ../donnees/open_meteo_toulouse.json")
    with open("../donnees/open_meteo_toulouse.json", encoding="utf-8") as fichier:
        en_direct = json.load(fichier)
    jours_recus = en_direct["daily"]["time"]

print("Premiers jours reçus :", jours_recus[:3])
````

`type(erreur).__name__` donne le **nom** du type d'erreur (par exemple `ConnectionError`, « erreur de connexion »).
La ligne `jours_recus = ...` est dans le `try` : si l'API répond quelque chose d'inattendu (sans la clé `daily`),
c'est aussi le plan B qui s'applique.

Avec du réseau, tu reçois les **vraies prévisions du jour**. Sans réseau, le plan B a pris le relais. Dans les deux
cas, la suite du programme a la **même structure** de données.

**Pour la suite du cours (et pour tous les exercices)**, on travaille sur la réponse **figée**
`open_meteo_toulouse.json`, enregistrée un matin d'octobre 2026 : elle ne change pas, donc les résultats sont les

## 10. Fusionner les prévisions avec l'historique

Notre historique (`meteo.csv`) est rangé **par ligne** (une liste de dictionnaires) ; l'API range **par colonne**.
Pour les fusionner, on transforme la réponse en liste de dictionnaires, avec une boucle sur les **positions**
(`range(len(...))`, module 05) :

````python
def previsions_en_releves(reponse, ville):
    """Réponse Open-Meteo (rangée par colonne) -> liste de relevés (un dictionnaire par jour)."""
    quotidien = reponse["daily"]
    releves = []
    for i in range(len(quotidien["time"])):
        releves.append({
            "date": quotidien["time"][i],
            "ville": ville,
            "temp_min": quotidien["temperature_2m_min"][i],
            "temp_max": quotidien["temperature_2m_max"][i],
            "pluie_mm": quotidien["precipitation_sum"][i],
            "source": "prévision",
        })
    return releves

previsions = previsions_en_releves(previsions_brutes, "Toulouse")
print(previsions[0])
````
