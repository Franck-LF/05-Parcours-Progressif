## Comment travailler

Chaque exercice reprend **un geste montré dans le cours**. Pour chacun :

1. Lis l'**objectif** et les **étapes**.
2. Remplace chaque `...` par ton code ; garde les noms de variables (et de fonctions) déjà écrits : la vérification en a besoin.
3. Exécute ta cellule (`Shift + Entrée`) et compare au **résultat attendu**.
4. Exécute la cellule **Vérification** juste en dessous : `✅ OK` = réussi ; sinon lis le message, il te dit quoi corriger.
5. Bloqué plus de 5 minutes ? Ouvre l'indice 💡, puis relis la section du cours indiquée (📖).

### Lire un échec de vérification

Quand une vérification échoue, Python affiche un long message rouge, surtout en anglais. **Ne lis que la
dernière ligne** :

```
AssertionError: moyenne doit valoir 21.5, tu as obtenu 25.0 : as-tu divisé par le nombre de jours ?
```

- `AssertionError` veut dire « **la vérification a échoué** » (ton code tourne, mais le résultat n'est pas encore le bon) ;
- le **message en français**, après les deux-points, te dit **quoi corriger** ;
- tout ce qui est au-dessus (en anglais, avec des numéros de ligne) peut être **ignoré**.

Si la dernière ligne commence par un autre mot (`NameError`, `TypeError`...), c'est **ta cellule** qui a un
problème : relis la section ⚠️ du cours qui parle de cette erreur.

Les caractères spéciaux (émojis comme ☀️ 🌧️ 💨, flèche →, symbole °) ne se tapent pas facilement : **copie-colle-les
depuis l'énoncé**.

Si un message de vérification parle d'`Ellipsis`, c'est le nom que Python donne à `...` : tu as laissé un `...`
quelque part.

Chaque exercice est **indépendant** : il redonne ses données. Tu peux donc les faire dans l'ordre que tu veux,
mais l'ordre proposé va du plus guidé au plus libre.

| Niveau | Ce que tu fais |
|---|---|
| **Niveau 1 — Imiter** | le même geste que le cours, presque à l'identique ; tu remplaces 1 ou 2 `...` |
| **Niveau 2 — Adapter** | le même geste dans un autre contexte, ou combiné avec un geste d'avant |
| **Niveau 3 — Transposer** | un petit problème décrit en français ; à toi de trouver les gestes |
| **🔁 Révisions espacées** | des gestes des modules précédents, pour ne pas les oublier |


**Pas d'Internet ici** : les exercices sur l'API lisent la réponse figée `../donnees/open_meteo_toulouse.json`,
exactement ce que `reponse.json()` aurait renvoyé. Les modules que tu crées sont écrits dans `sorties/`.
